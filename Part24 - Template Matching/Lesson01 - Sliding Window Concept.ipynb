# Lesson 01 — Sliding Window Concept

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img      = cv2.imread('sample.jpg')
gray     = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
# Crop a template from the image
template = gray[100:160, 200:280]  # adjust to a distinctive region
tH, tW   = template.shape

# TM_CCOEFF_NORMED: best general-purpose method
result   = cv2.matchTemplate(gray, template, cv2.TM_CCOEFF_NORMED)
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

vis = img.copy()
cv2.rectangle(vis, max_loc, (max_loc[0]+tW, max_loc[1]+tH), (0,255,0), 3)
cv2.rectangle(vis, (200,100),(280,160),(0,0,255),2)  # original location

fig,axes=plt.subplots(1,3,figsize=(18,5))
axes[0].imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)); axes[0].set_title('Image (red=template crop)'); axes[0].axis('off')
axes[1].imshow(template,cmap='gray');                 axes[1].set_title('Template'); axes[1].axis('off')
axes[2].imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB)); axes[2].set_title(f'Match found (score={max_val:.3f})'); axes[2].axis('off')
plt.show()
print(f"Best match at {max_loc}, score={max_val:.3f} (1.0=perfect)")
```

## Key Takeaway
Template matching slides the template over the image and computes similarity at each position.
TM_CCOEFF_NORMED returns scores in [-1,1]. Score > 0.8 usually indicates a good match.
