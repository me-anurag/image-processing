# Lesson 02 — The 6 Matching Methods

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img      = cv2.imread('sample.jpg')
gray     = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
template = gray[80:160, 150:250]
tH, tW   = template.shape

methods = [
    ('TM_SQDIFF',        cv2.TM_SQDIFF,        False),  # low=good
    ('TM_SQDIFF_NORMED', cv2.TM_SQDIFF_NORMED, False),
    ('TM_CCORR',         cv2.TM_CCORR,         True),   # high=good
    ('TM_CCORR_NORMED',  cv2.TM_CCORR_NORMED,  True),
    ('TM_CCOEFF',        cv2.TM_CCOEFF,        True),
    ('TM_CCOEFF_NORMED', cv2.TM_CCOEFF_NORMED, True),   # best choice
]

fig, axes = plt.subplots(2,6,figsize=(30,8))
for col, (name, method, max_is_good) in enumerate(methods):
    res = cv2.matchTemplate(gray, template, method)
    if max_is_good:
        _, _, _, loc = cv2.minMaxLoc(res)
        loc = cv2.minMaxLoc(res)[3]
    else:
        loc = cv2.minMaxLoc(res)[2]
    vis = img.copy()
    cv2.rectangle(vis, loc, (loc[0]+tW,loc[1]+tH), (0,255,0), 3)
    axes[0,col].imshow(cv2.normalize(res,None,0,255,cv2.NORM_MINMAX).astype('uint8'),cmap='hot')
    axes[0,col].set_title(name,fontsize=8); axes[0,col].axis('off')
    axes[1,col].imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB)); axes[1,col].axis('off')
plt.suptitle('6 matching methods — TM_CCOEFF_NORMED is almost always the right choice',fontsize=11)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Always use TM_CCOEFF_NORMED. It's normalized (insensitive to brightness) and max-is-good (intuitive).
