# Lesson 04 — Multiple Instance Detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img      = cv2.imread('sample.jpg')
gray     = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
template = gray[80:150, 100:200]
tH, tW   = template.shape

result = cv2.matchTemplate(gray, template, cv2.TM_CCOEFF_NORMED)

# Threshold to find ALL locations above 0.7
threshold = 0.75
locs = np.where(result >= threshold)

# Non-maximum suppression: remove overlapping boxes
boxes = [[int(x),int(y),int(x+tW),int(y+tH)] for x,y in zip(locs[1],locs[0])]

vis = img.copy()
for (x1,y1,x2,y2) in boxes:
    cv2.rectangle(vis,(x1,y1),(x2,y2),(0,255,0),2)

plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB))
plt.title(f'{len(boxes)} instances found (threshold={threshold})'); plt.axis('off'); plt.show()
```

## Key Takeaway
For multiple instances: threshold result map instead of just taking max location.
Apply NMS if boxes overlap heavily.
