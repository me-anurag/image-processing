# Lesson 05 — Limitations of Template Matching

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img      = cv2.imread('sample.jpg')
gray     = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
template = gray[80:160, 150:250]

# Test: template matching fails on rotated template
M = cv2.getRotationMatrix2D((template.shape[1]//2,template.shape[0]//2), 30, 1.0)
rot_template = cv2.warpAffine(template, M, (template.shape[1],template.shape[0]))

res_orig = cv2.matchTemplate(gray, template,     cv2.TM_CCOEFF_NORMED)
res_rot  = cv2.matchTemplate(gray, rot_template, cv2.TM_CCOEFF_NORMED)

_,orig_score,_,_ = cv2.minMaxLoc(res_orig)
_,rot_score,_,_  = cv2.minMaxLoc(res_rot)

print(f"Original template score:         {orig_score:.3f}  (high = good match)")
print(f"Rotated  template score:         {rot_score:.3f}  (much worse!)")
print()
print("LIMITATIONS OF TEMPLATE MATCHING:")
print("  ✗ Fails on rotation")
print("  ✗ Fails on scale change (unless multi-scale)")
print("  ✗ Fails on perspective change")
print("  ✗ Sensitive to lighting change (use NORMED methods)")
print("  ✓ Perfect for: logos at known size, UI elements, stamps, watermarks")
print()")
print("Use FEATURE MATCHING (ORB/SIFT) when rotation/scale invariance is needed.")
```

## Key Takeaway
Template matching works perfectly for exact appearance at the same scale/orientation.
For any viewpoint change, use feature matching + homography instead.
