# Lesson 03 — Reading, Displaying, and Saving Images

## Why This Lesson
Every pipeline starts by loading an image and ends by saving one. Get this wrong and nothing works.

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# READ
img = cv2.imread('sample.jpg')           # BGR
img_gray = cv2.imread('sample.jpg', 0)   # grayscale directly

# DISPLAY with matplotlib (always convert BGR→RGB first)
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Color'); plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('Grayscale'); plt.axis('off')
plt.show()

# SAVE
cv2.imwrite('output.jpg', img)           # save as JPEG
cv2.imwrite('output.png', img)           # save as PNG (lossless)

print("Saved successfully")
```

## Key Takeaway
Always convert BGR→RGB before displaying with matplotlib. `cv2.imshow` handles BGR natively but only works in `.py` files, not notebooks.
