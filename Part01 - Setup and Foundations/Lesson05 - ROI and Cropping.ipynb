# Lesson 05 — ROI and Cropping

## Why This Lesson
You rarely need to process an entire image. Cropping to the region of interest (ROI) speeds everything up and reduces noise from irrelevant areas.

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Crop: img[y1:y2, x1:x2]
h, w = img.shape[:2]

# Crop center region
cx, cy = w//2, h//2
roi = img[cy-100:cy+100, cx-100:cx+100]

# Crop top-left quarter
top_left = img[0:h//2, 0:w//2]

# Paste ROI back at a different location
result = img.copy()
result[0:roi.shape[0], 0:roi.shape[1]] = roi

fig, axes = plt.subplots(1, 3, figsize=(15,5))
for ax, im, t in zip(axes, [img[:,:,::-1], roi[:,:,::-1], result[:,:,::-1]],
                           ['Original', 'ROI (center)', 'ROI pasted top-left']):
    ax.imshow(im); ax.set_title(t); ax.axis('off')
plt.show()
```

Cropping is just NumPy slicing: `img[y1:y2, x1:x2]`. It's instant — no copy is made until you modify it.
