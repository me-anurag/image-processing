# Lesson 01 — Installing OpenCV, NumPy, and Matplotlib

## Why This Lesson
You cannot process images without tools. This lesson gets your environment ready and verifies every library works correctly.

## Concepts
- OpenCV (`cv2`) — the main image processing library
- NumPy — images are NumPy arrays; this is not optional knowledge
- Matplotlib — for displaying images inline in notebooks

## Installation
```bash
pip install opencv-python numpy matplotlib
```

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Verify versions
print("OpenCV:", cv2.__version__)
print("NumPy:", np.__version__)

# Create a simple test image from scratch
img = np.zeros((300, 300, 3), dtype=np.uint8)
img[100:200, 100:200] = [0, 255, 0]  # green square

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Your first image — a green square")
plt.axis('off')
plt.show()
```

## What Just Happened
You created a 300×300 black image (all zeros), painted a green square in the middle by directly setting pixel values, and displayed it. This is the foundation of everything.

## Key Takeaway
An image is just a NumPy array. Every pixel is a number. That's all it ever is.
