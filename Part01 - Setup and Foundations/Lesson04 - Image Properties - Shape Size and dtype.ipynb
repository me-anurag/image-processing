# Lesson 04 — Image Properties: Shape, Size, and dtype

## Why This Lesson
Knowing how to interrogate an image — before touching it — prevents crashes and wrong outputs.

## Code
```python
import cv2
import numpy as np

img = cv2.imread('sample.jpg')

# Core properties
height, width, channels = img.shape
print(f"Width: {width}px, Height: {height}px, Channels: {channels}")
print(f"dtype: {img.dtype}")
print(f"Total pixels: {img.size // channels}")
print(f"Array size in memory: {img.nbytes / 1024:.1f} KB")

# Pixel value statistics
print(f"Min: {img.min()}, Max: {img.max()}, Mean: {img.mean():.1f}")

# Check if grayscale or color
if len(img.shape) == 2:
    print("Grayscale image")
else:
    print(f"Color image with {img.shape[2]} channels")
```
