# Lesson 02 — What is an Image: Pixels, Channels, and dtype

## Why This Lesson
Before processing images you must understand what they ARE. An image is not magic — it's a grid of numbers. Every confusion you'll ever have traces back to not knowing this deeply enough.

## Concepts
- **Pixel**: one cell in the grid, holds a color value
- **Channel**: one color component (Blue, Green, Red in OpenCV)
- **dtype**: the data type — `uint8` means values 0–255; `float32` means 0.0–1.0

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load an image
img = cv2.imread('sample.jpg')  # loads as BGR, shape: (H, W, 3)

print("Shape:", img.shape)       # (height, width, channels)
print("dtype:", img.dtype)       # uint8
print("Min pixel:", img.min())   # 0
print("Max pixel:", img.max())   # 255

# Access a single pixel
pixel = img[100, 150]            # row 100, col 150
print("Pixel at (100,150):", pixel)  # [B, G, R]

# Split channels
b, g, r = cv2.split(img)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, chan, name in zip(axes, [img[:,:,::-1], b, g, r], ['Original','Blue','Green','Red']):
    ax.imshow(chan, cmap='gray' if name != 'Original' else None)
    ax.set_title(name); ax.axis('off')
plt.show()
```

## Key Takeaway
Shape is (Height, Width, Channels). dtype is almost always uint8. OpenCV stores channels as BGR not RGB. These three facts will save you hours of confusion.
