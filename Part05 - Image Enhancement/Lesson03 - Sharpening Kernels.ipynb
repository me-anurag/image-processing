# Lesson 03 — Sharpening Kernels

## Why
Blurry images lose high-frequency detail. Sharpening kernels amplify edges and fine details. Understanding how they work lets you build any custom enhancement.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Standard sharpen kernel
k_sharpen = np.array([[ 0, -1,  0],
                       [-1,  5, -1],
                       [ 0, -1,  0]], dtype=np.float32)

# Strong sharpen (emphasizes all neighbors)
k_strong  = np.array([[-1, -1, -1],
                       [-1,  9, -1],
                       [-1, -1, -1]], dtype=np.float32)

# Unsharp masking — best method
def unsharp_mask(img, sigma=1.5, strength=1.5):
    blurred = cv2.GaussianBlur(img, (0, 0), sigma)
    sharpened = cv2.addWeighted(img, 1 + strength, blurred, -strength, 0)
    return np.clip(sharpened, 0, 255).astype(np.uint8)

sharpened = cv2.filter2D(img, -1, k_sharpen)
strong    = cv2.filter2D(img, -1, k_strong)
unsharp   = unsharp_mask(img, sigma=1.5, strength=1.2)

fig, axes = plt.subplots(1, 4, figsize=(22, 6))
for ax, im, t in zip(axes,
    [img, sharpened, strong, unsharp],
    ['Original', 'Sharpen kernel', 'Strong sharpen', 'Unsharp Mask (best)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t, fontsize=12); ax.axis('off')
plt.tight_layout(); plt.show()

# Demonstrate: apply to a deliberately blurred image
blurry = cv2.GaussianBlur(img, (7, 7), 0)
recovered = unsharp_mask(blurry, sigma=2.0, strength=2.0)
fig, axes = plt.subplots(1,3,figsize=(18,5))
for ax, im, t in zip(axes, [img, blurry, recovered], ['Sharp original','Blurred','Unsharp mask applied']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Unsharp masking = original + scaled(original − blurred). It subtracts a blurred version (low frequencies) from the original, which amplifies the difference (high frequencies = edges/detail).
