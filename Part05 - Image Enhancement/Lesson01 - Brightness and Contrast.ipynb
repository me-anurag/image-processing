# Lesson 01 — Brightness and Contrast

## The Formula
`output = alpha × input + beta`

| Parameter | Effect |
|---|---|
| alpha > 1 | Increase contrast |
| alpha < 1 | Decrease contrast |
| beta > 0  | Increase brightness |
| beta < 0  | Decrease brightness |

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

def adjust(img, alpha=1.0, beta=0):
    # np.clip prevents values wrapping around (critical!)
    return np.clip(alpha * img.astype(np.float32) + beta, 0, 255).astype(np.uint8)

samples = {
    'Original'     : adjust(img, 1.0,   0),
    'Brighter'     : adjust(img, 1.0,  70),
    'Darker'       : adjust(img, 1.0, -70),
    'High Contrast': adjust(img, 2.0, -50),
    'Low Contrast' : adjust(img, 0.4,  80),
    'Negative'     : adjust(img, -1.0, 255),
}

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, (name, im) in zip(axes.flat, samples.items()):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(name, fontsize=13); ax.axis('off')
plt.suptitle('alpha × pixel + beta   —   alpha=contrast, beta=brightness', fontsize=13)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Always use `np.clip(result, 0, 255).astype(np.uint8)`. Without clip, values overflow and wrap around causing severe artifacts.
