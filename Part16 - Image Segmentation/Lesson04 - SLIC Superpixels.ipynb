# Lesson 04 — SLIC Superpixels

## Why
Superpixels are groups of pixels that belong together perceptually.
Instead of processing individual pixels, you process superpixels — orders of magnitude fewer units.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

try:
    from skimage.segmentation import slic, mark_boundaries
    from skimage.util import img_as_float

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_float = img_as_float(img_rgb)

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    axes[0].imshow(img_rgb); axes[0].set_title('Original'); axes[0].axis('off')

    for ax, n_segs in zip(axes[1:], [100, 400]):
        segments = slic(img_float, n_segments=n_segs, compactness=10, sigma=1)
        boundary = mark_boundaries(img_float, segments)
        ax.imshow(boundary)
        ax.set_title(f'SLIC: {n_segs} superpixels'); ax.axis('off')

    plt.suptitle('Superpixels: perceptually uniform clusters of pixels', fontsize=12)
    plt.tight_layout(); plt.show()
    print(f"Unique superpixels: {len(np.unique(segments))}")

except ImportError:
    print("Install scikit-image: pip install scikit-image")
    print("Then rerun this cell.")
    # OpenCV alternative: SEEDS superpixels
    seeds = cv2.ximgproc.createSuperpixelSEEDS(img.shape[1], img.shape[0], img.shape[2], 200, 8)
    seeds.iterate(img, 10)
    labels = seeds.getLabels()
    mask   = seeds.getLabelContourMask()
    vis    = img.copy()
    vis[mask==255] = [0,255,0]
    plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    plt.title('SEEDS Superpixels (OpenCV ximgproc)'); plt.axis('off'); plt.show()
```

## Key Takeaway
Superpixels reduce computational complexity while respecting object boundaries.
Used in: image classification preprocessing, saliency detection, weakly-supervised segmentation.
