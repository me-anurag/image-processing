# Lesson 03 — Mean Shift Segmentation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# pyrMeanShiftFiltering: spatial radius, color radius, pyramid levels
result_coarse = cv2.pyrMeanShiftFiltering(img, sp=20, sr=40)
result_fine   = cv2.pyrMeanShiftFiltering(img, sp=10, sr=20)
result_heavy  = cv2.pyrMeanShiftFiltering(img, sp=40, sr=80)

fig, axes = plt.subplots(1, 4, figsize=(22,5))
for ax, im, t in zip(axes,
    [img, result_fine, result_coarse, result_heavy],
    ['Original', 'Fine (sp=10,sr=20)', 'Medium (sp=20,sr=40)', 'Heavy (sp=40,sr=80)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.suptitle('Mean shift: flattens regions to uniform color — cartoon-like', fontsize=12)
plt.show()
```

## Key Takeaway
sp = spatial radius (how far to look for similar pixels).
sr = color radius (how similar in color pixels must be to merge).
Higher values = more segmentation, fewer regions, more cartoon-like.
