# Lesson 03 — Contour Hierarchy: Parent and Child

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create nested shapes: outer ring + inner shape (donut)
canvas = np.zeros((400, 600), dtype=np.uint8)
cv2.circle(canvas, (200, 200), 120, 255, -1)  # outer circle
cv2.circle(canvas, (200, 200),  60, 0,   -1)  # hole inside
cv2.rectangle(canvas, (350, 100), (550, 300), 255, -1)  # outer rect
cv2.rectangle(canvas, (390, 140), (510, 260), 0,   -1)  # hole inside

# RETR_CCOMP gives 2-level hierarchy (outer=0, inner=1)
contours_ext, hier_ext = cv2.findContours(canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contours_all, hier_all = cv2.findContours(canvas, cv2.RETR_CCOMP,    cv2.CHAIN_APPROX_SIMPLE)

print(f"RETR_EXTERNAL: {len(contours_ext)} contours (outer only)")
print(f"RETR_CCOMP:    {len(contours_all)} contours (outer + holes)")
print("Hierarchy [next, prev, first_child, parent]:")
for i, h in enumerate(hier_all[0]):
    print(f"  Contour {i}: {h}  {'(outer)' if h[3]==-1 else '(hole)'}")

vis = cv2.cvtColor(canvas, cv2.COLOR_GRAY2BGR)
for i, c in enumerate(contours_all):
    is_hole = hier_all[0][i][3] != -1
    color   = (0,0,255) if is_hole else (0,255,0)
    cv2.drawContours(vis, [c], -1, color, 3)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.imshow(canvas, cmap='gray'); plt.title('Shapes with holes')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); plt.title('Green=outer, Red=holes')
plt.show()
```

## Key Takeaway
RETR_EXTERNAL returns only outermost contours — use for simple counting.
RETR_CCOMP returns two levels — use when holes matter (donuts, letters with holes like O, B, D).
