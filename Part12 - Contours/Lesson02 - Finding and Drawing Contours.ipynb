# Lesson 02 — Finding and Drawing Contours

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Clean up
se     = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
binary = cv2.morphologyEx(binary, cv2.MORPH_OPEN, se)

contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
print(f"Total contours found: {len(contours)}")

# Draw ALL contours
all_contours = img.copy()
cv2.drawContours(all_contours, contours, -1, (0,255,0), 2)

# Draw only LARGE contours (filter by area)
min_area   = 500
large_cnts = [c for c in contours if cv2.contourArea(c) > min_area]
large_vis  = img.copy()
for i, c in enumerate(large_cnts):
    color = tuple(np.random.randint(50, 255, 3).tolist())
    cv2.drawContours(large_vis, [c], -1, color, 2)
    x,y,w,h = cv2.boundingRect(c)
    cv2.putText(large_vis, f'{cv2.contourArea(c):.0f}', (x,y-5),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 1)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
for ax, im, t in zip(axes,
    [img, all_contours, large_vis],
    ['Original', f'All {len(contours)} contours', f'Large only ({len(large_cnts)}, area>{min_area})']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Filter by `cv2.contourArea(c) > min_area` to remove noise contours.
Each contour is a numpy array of (N,1,2) shape — N points, each (x,y).
