# Lesson 04 — Contour Approximation: Douglas-Peucker

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)

if contours:
    c = max(contours, key=cv2.contourArea)
    print(f"Original contour: {len(c)} points")

    epsilons = [0.001, 0.01, 0.02, 0.05]
    fig, axes = plt.subplots(1, 5, figsize=(26, 5))
    vis0 = np.zeros((*binary.shape, 3), dtype=np.uint8)
    cv2.drawContours(vis0, [c], -1, (0,255,0), 2)
    axes[0].imshow(vis0); axes[0].set_title(f'Original
{len(c)} pts'); axes[0].axis('off')

    peri = cv2.arcLength(c, True)
    for ax, eps in zip(axes[1:], epsilons):
        approx = cv2.approxPolyDP(c, eps * peri, True)
        vis = np.zeros((*binary.shape, 3), dtype=np.uint8)
        cv2.drawContours(vis, [approx], -1, (0,255,255), 2)
        for pt in approx:
            cv2.circle(vis, tuple(pt[0]), 5, (0,0,255), -1)
        ax.imshow(vis)
        ax.set_title(f'ε={eps}
{len(approx)} pts'); ax.axis('off')
    plt.suptitle('Higher epsilon = fewer points = simpler shape', fontsize=12)
    plt.show()
```

## Key Takeaway
`approxPolyDP(c, epsilon * perimeter, True)` — epsilon is the fraction of perimeter allowed as error.
epsilon=0.02 is the magic value for detecting quadrilaterals (documents, cards).
epsilon=0.04 for pentagons. epsilon=0.1 for rough approximations.
