# Lesson 01 — What is a Contour

## Why This Lesson
A contour is a curve joining all continuous points along a boundary with the same color/intensity.
Contours are how OpenCV represents object boundaries — and everything you can measure about a shape.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a simple binary image with clear shapes
canvas = np.zeros((400, 600), dtype=np.uint8)
cv2.rectangle(canvas, (50,  50),  (200, 180), 255, -1)
cv2.circle(canvas,    (380, 120), 80, 255, -1)
cv2.ellipse(canvas,   (130, 300), (90,50), 0, 0, 360, 255, -1)

# Find contours
contours, hierarchy = cv2.findContours(canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

print(f"Found {len(contours)} contours")
for i, c in enumerate(contours):
    print(f"  Contour {i}: {len(c)} points,  area={cv2.contourArea(c):.0f}")

# Draw contours
vis = cv2.cvtColor(canvas, cv2.COLOR_GRAY2BGR)
colors = [(0,255,0), (255,0,0), (0,0,255)]
for i, (c, col) in enumerate(zip(contours, colors)):
    cv2.drawContours(vis, [c], -1, col, 3)
    M  = cv2.moments(c)
    cx = int(M['m10']/(M['m00']+1e-6))
    cy = int(M['m01']/(M['m00']+1e-6))
    cv2.putText(vis, str(i), (cx-10,cy+10), cv2.FONT_HERSHEY_SIMPLEX, 1, col, 2)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.imshow(canvas, cmap='gray'); plt.title('Binary image'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); plt.title('Contours drawn'); plt.axis('off')
plt.show()
```

## Key Takeaway
Contours need a binary image. Always threshold first.
`cv2.RETR_EXTERNAL` returns only outer contours.
`cv2.CHAIN_APPROX_SIMPLE` compresses straight lines to endpoints (saves memory).
