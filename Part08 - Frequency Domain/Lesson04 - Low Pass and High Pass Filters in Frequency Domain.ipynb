# Lesson 04 — Low-Pass and High-Pass Filters in Frequency Domain

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)

dft         = cv2.dft(gray, flags=cv2.DFT_COMPLEX_OUTPUT)
dft_shifted = np.fft.fftshift(dft)
h, w = gray.shape
cx, cy = w//2, h//2

# Compare different filter radii
radii = [10, 30, 80, 150]
fig, axes = plt.subplots(2, 4, figsize=(22, 10))

for col, r in enumerate(radii):
    mask = np.zeros((h, w, 2), np.uint8)
    cv2.circle(mask, (cx,cy), r, (1,1), -1)
    back   = np.fft.ifftshift(dft_shifted * mask)
    result = cv2.idft(back, flags=cv2.DFT_SCALE | cv2.DFT_REAL_OUTPUT)
    result = cv2.normalize(result, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    mag_vis = dft_shifted[:,:,0].copy()
    cv2.circle(mag_vis.astype(np.uint8), (cx,cy), r, 255, 2)

    axes[0,col].imshow(result, cmap='gray'); axes[0,col].set_title(f'Low-pass r={r}'); axes[0,col].axis('off')
    axes[1,col].imshow(np.log1p(cv2.magnitude(dft_shifted[:,:,0], dft_shifted[:,:,1])), cmap='hot')
    cv2.circle
    axes[1,col].add_patch(plt.Circle((cx,cy), r, fill=False, color='white', linewidth=2))
    axes[1,col].set_title(f'Spectrum (circle=cutoff)'); axes[1,col].axis('off')

plt.suptitle('Smaller radius = more blur (fewer frequencies kept)', fontsize=12)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Low-pass = keep center frequencies = smooth/blur.
High-pass = keep outer frequencies = edges only.
The cutoff radius in frequency domain corresponds to the sigma in Gaussian blur spatially.
