# Lesson 01 — Fourier Transform Intuition

## Why This Lesson
Every image is a sum of sine waves at different frequencies.
Low frequencies = smooth regions. High frequencies = edges and fine details.
Seeing in frequency is a completely different superpower from working in spatial domain.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)

# Compute DFT
dft     = cv2.dft(gray, flags=cv2.DFT_COMPLEX_OUTPUT)
dft_shifted = np.fft.fftshift(dft)  # shift zero-frequency to center

# Magnitude spectrum (log scale for visibility)
magnitude = cv2.magnitude(dft_shifted[:,:,0], dft_shifted[:,:,1])
magnitude_log = np.log1p(magnitude)
magnitude_log = cv2.normalize(magnitude_log, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(gray, cmap='gray'); axes[0].set_title('Spatial domain (what you see)'); axes[0].axis('off')
axes[1].imshow(magnitude_log, cmap='hot'); axes[1].set_title('Frequency domain — magnitude spectrum\nCenter=low freq, Edges=high freq'); axes[1].axis('off')
plt.tight_layout(); plt.show()

print("Center bright spot = DC component (average brightness)")
print("Bright lines at angles = strong periodic structures (walls, grids) at that angle")
```

## Key Takeaway
Center of spectrum = low frequencies (overall brightness, gradients).
Outer edges = high frequencies (edges, noise, fine texture).
Removing outer frequencies = blur. Removing center = edge detection.
