# Lesson 02 — Per-Channel Histograms

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5))
ax1.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax1.axis('off'); ax1.set_title('Image')

for i, (color, name) in enumerate(zip(('blue','green','red'), ('Blue','Green','Red'))):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    ax2.plot(hist, color=color, label=name, alpha=0.8, linewidth=1.2)
    ax2.fill_between(range(256), hist.flatten(), alpha=0.15, color=color)

ax2.set_xlim([0, 256]); ax2.legend()
ax2.set_title('Per-Channel Histogram — tells you the color cast')
ax2.set_xlabel('Pixel Value'); ax2.set_ylabel('Frequency')
plt.tight_layout(); plt.show()

# Diagnosis
b, g, r = cv2.split(img)
print(f"Channel means  B={b.mean():.0f}  G={g.mean():.0f}  R={r.mean():.0f}")
dominant = ['Blue','Green','Red'][np.argmax([b.mean(), g.mean(), r.mean()])]
print(f"Dominant channel: {dominant}")
```

## Key Takeaway
If the Red histogram is shifted right → warm/orange cast. If Blue is dominant → cool/blue cast. Histograms diagnose white balance problems instantly.
