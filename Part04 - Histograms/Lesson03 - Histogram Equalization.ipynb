# Lesson 03 — Histogram Equalization

## Why This Lesson
Equalization spreads a bunched histogram across the full 0–255 range — maximizing contrast automatically. It's the oldest and simplest auto-enhancement.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

equalized = cv2.equalizeHist(gray)

fig, axes = plt.subplots(2, 2, figsize=(13, 10))
axes[0,0].imshow(gray,       cmap='gray'); axes[0,0].set_title('Original');   axes[0,0].axis('off')
axes[0,1].imshow(equalized,  cmap='gray'); axes[0,1].set_title('Equalized');  axes[0,1].axis('off')
axes[1,0].hist(gray.flatten(),      256, [0,256], color='steelblue', edgecolor='none')
axes[1,0].set_title('Original histogram — bunched')
axes[1,1].hist(equalized.flatten(), 256, [0,256], color='steelblue', edgecolor='none')
axes[1,1].set_title('Equalized histogram — spread out')
plt.tight_layout(); plt.show()
```

## Key Takeaway
Equalization is global — it processes the entire image with one uniform mapping. This causes problems where one part of the image is dark and another is bright. That's exactly why CLAHE was invented.
