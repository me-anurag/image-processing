# Lesson 01 — What is a Histogram

## Why This Lesson
A histogram tells you everything about an image's exposure and tone distribution before you even look at it. A CV engineer reads histograms the way a doctor reads an X-ray.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

hist = cv2.calcHist([gray], [0], None, [256], [0, 256])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.imshow(gray, cmap='gray'); ax1.set_title('Image'); ax1.axis('off')

ax2.plot(hist, color='black', linewidth=0.8)
ax2.fill_between(range(256), hist.flatten(), alpha=0.3, color='gray')
ax2.set_xlim([0, 256])
ax2.set_xlabel('Pixel Intensity (0=black  255=white)')
ax2.set_ylabel('Number of pixels')
ax2.set_title('Intensity Histogram')
ax2.axvline(x=gray.mean(), color='red', linestyle='--', label=f'Mean = {gray.mean():.0f}')
ax2.legend()
plt.tight_layout(); plt.show()

mean = gray.mean()
if mean < 80:
    print(f"Diagnosis: UNDEREXPOSED (mean={mean:.0f})")
elif mean > 180:
    print(f"Diagnosis: OVEREXPOSED  (mean={mean:.0f})")
else:
    print(f"Diagnosis: NORMAL       (mean={mean:.0f})")
```

## Key Takeaway
Histogram bunched left = dark/underexposed. Bunched right = bright/overexposed. Spread evenly across 0–255 = well-exposed with good contrast.
