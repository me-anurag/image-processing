# Lesson 04 — CLAHE: Adaptive Equalization

## Why This Lesson
CLAHE splits the image into small tiles and equalizes each independently. The result handles uneven lighting that global equalization destroys.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Apply CLAHE on the L channel of LAB — NEVER directly on BGR
lab    = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
l, a, b = cv2.split(lab)

clahe   = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
l_clahe = clahe.apply(l)

result = cv2.cvtColor(cv2.merge([l_clahe, a, b]), cv2.COLOR_LAB2BGR)

# Compare: original vs global eq vs CLAHE
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
global_eq = cv2.cvtColor(cv2.equalizeHist(gray), cv2.COLOR_GRAY2BGR)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
for ax, im, t in zip(axes,
    [img, global_eq, result],
    ['Original', 'Global Equalization', 'CLAHE (better)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t, fontsize=13); ax.axis('off')
plt.suptitle('CLAHE handles uneven lighting that global equalization cannot', fontsize=12)
plt.tight_layout(); plt.show()

# clipLimit experiment
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, clip in zip(axes, [0.5, 2.0, 5.0, 15.0]):
    c  = cv2.createCLAHE(clipLimit=clip, tileGridSize=(8,8))
    lc = c.apply(l)
    r  = cv2.cvtColor(cv2.merge([lc, a, b]), cv2.COLOR_LAB2BGR)
    ax.imshow(cv2.cvtColor(r, cv2.COLOR_BGR2RGB))
    ax.set_title(f'clipLimit={clip}'); ax.axis('off')
plt.suptitle('Higher clipLimit = more aggressive — find the sweet spot', fontsize=12)
plt.show()
```

## Key Takeaway
Always apply CLAHE on the L channel of LAB. `clipLimit=2.0, tileGridSize=(8,8)` is the universal safe default. Go higher only for severely degraded images.
