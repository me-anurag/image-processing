# Lesson 03 — Opening and Closing

## Why This Lesson
Opening and closing combine erosion and dilation to achieve specific cleaning effects
without the side-effect of permanently changing object size.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

se = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9,9))

# Opening = Erosion then Dilation → removes small white noise dots
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, se)

# Closing = Dilation then Erosion → fills small holes in white objects
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, se)

# Add artificial noise/holes to demonstrate
noisy = binary.copy()
for _ in range(300):
    x, y = np.random.randint(0, binary.shape[1]), np.random.randint(0, binary.shape[0])
    cv2.circle(noisy, (x,y), 3, 255, -1)  # noise dots

holey = binary.copy()
for _ in range(200):
    x, y = np.random.randint(0, binary.shape[1]), np.random.randint(0, binary.shape[0])
    cv2.circle(holey, (x,y), 5, 0, -1)   # holes

opened_noisy = cv2.morphologyEx(noisy, cv2.MORPH_OPEN,  se)
closed_holey = cv2.morphologyEx(holey, cv2.MORPH_CLOSE, se)

fig, axes = plt.subplots(2, 2, figsize=(14, 11))
axes[0,0].imshow(noisy,        cmap='gray'); axes[0,0].set_title('Noisy (random dots)'); axes[0,0].axis('off')
axes[0,1].imshow(opened_noisy, cmap='gray'); axes[0,1].set_title('After OPENING — dots removed'); axes[0,1].axis('off')
axes[1,0].imshow(holey,        cmap='gray'); axes[1,0].set_title('Holey (random holes)'); axes[1,0].axis('off')
axes[1,1].imshow(closed_holey, cmap='gray'); axes[1,1].set_title('After CLOSING — holes filled'); axes[1,1].axis('off')
plt.suptitle('Opening removes small bright noise. Closing fills small dark holes.', fontsize=12)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Opening = remove noise dots (smaller than SE).
Closing = fill holes (smaller than SE).
