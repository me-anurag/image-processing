# Lesson 04 — Morphological Gradient, Top Hat, and Black Hat

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

se = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9,9))

# Gradient = Dilation - Erosion → gives outline/border of objects
gradient = cv2.morphologyEx(gray, cv2.MORPH_GRADIENT, se)

# Top Hat = Original - Opening → extracts bright regions smaller than SE
tophat   = cv2.morphologyEx(gray, cv2.MORPH_TOPHAT,   se)

# Black Hat = Closing - Original → extracts dark regions smaller than SE
blackhat = cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, se)

fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, im, t in zip(axes,
    [gray, gradient, tophat, blackhat],
    ['Original', 'Gradient (object outlines)', 'Top Hat (bright details)', 'Black Hat (dark details)']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.suptitle('Advanced morphological operations', fontsize=12)
plt.show()

# Practical: Top Hat for shadow removal from documents
# Top Hat removes uneven background (shadows), keeping only local bright features (text)
enhanced = cv2.add(gray, tophat)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.imshow(gray,     cmap='gray'); plt.title('With shadow');         plt.axis('off')
plt.subplot(1,2,2); plt.imshow(enhanced, cmap='gray'); plt.title('Top Hat shadow fix'); plt.axis('off')
plt.show()
```

## Key Takeaway
Top Hat extracts bright objects on a dark uneven background (great for text on shaded documents).
Black Hat extracts dark objects on bright background.
Gradient gives clean object outlines in one operation.
