# Lesson 01 — Structuring Elements

## Why This Lesson
The structuring element (SE) is the kernel in morphological operations.
Its shape and size determine what gets eroded or dilated.
Choosing the right SE is as important as choosing the right operation.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Three SE shapes
rect    = cv2.getStructuringElement(cv2.MORPH_RECT,    (15,15))
ellipse = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (15,15))
cross   = cv2.getStructuringElement(cv2.MORPH_CROSS,   (15,15))

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, se, t in zip(axes, [rect, ellipse, cross],
    ['RECT (square)', 'ELLIPSE (circle)', 'CROSS (plus)']):
    ax.imshow(se*255, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.suptitle('Structuring Elements — shape controls what morphology "sees"', fontsize=12)
plt.show()

# SE size comparison — effect on dilation
img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

fig, axes = plt.subplots(1, 4, figsize=(22, 5))
axes[0].imshow(binary, cmap='gray'); axes[0].set_title('Binary input'); axes[0].axis('off')
for ax, size in zip(axes[1:], [3, 9, 21]):
    se     = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (size, size))
    result = cv2.dilate(binary, se, iterations=1)
    ax.imshow(result, cmap='gray'); ax.set_title(f'Dilate SE={size}×{size}'); ax.axis('off')
plt.show()
```

## Key Takeaway
ELLIPSE SE for natural objects (cells, coins, rounded shapes).
RECT SE for text and man-made objects.
CROSS SE for thin linear structures.
