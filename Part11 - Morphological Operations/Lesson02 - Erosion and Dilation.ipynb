# Lesson 02 — Erosion and Dilation

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

se = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7,7))

eroded  = cv2.erode(binary,  se, iterations=1)
dilated = cv2.dilate(binary, se, iterations=1)
eroded3 = cv2.erode(binary,  se, iterations=3)
dilated3= cv2.dilate(binary, se, iterations=3)

fig, axes = plt.subplots(2, 3, figsize=(18, 11))
for ax, im, t in zip(axes.flat,
    [binary, eroded, dilated, binary, eroded3, dilated3],
    ['Original', 'Erode 1x (shrink)', 'Dilate 1x (expand)',
     'Original', 'Erode 3x (thin)', 'Dilate 3x (thick)']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.suptitle('Erosion shrinks white regions. Dilation expands them.', fontsize=12)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Erosion shrinks white regions (removes small white noise, separates touching objects).
Dilation expands white regions (fills small holes, connects nearby blobs).
