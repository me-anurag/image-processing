# Lesson 01 — Gradient: The Math Behind Edges

## Why This Lesson
An edge is where pixel values change rapidly. The gradient measures that rate of change.
Understanding gradient is understanding every edge detector that follows.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)

# Manual gradient using finite differences
grad_x_manual = np.diff(gray, axis=1)  # horizontal differences
grad_y_manual = np.diff(gray, axis=0)  # vertical differences

# OpenCV Sobel (better — uses 3x3 kernel, noise-resistant)
grad_x = cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3)  # X direction
grad_y = cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)  # Y direction

# Magnitude: sqrt(Gx^2 + Gy^2)
magnitude = cv2.magnitude(grad_x, grad_y)
magnitude = cv2.normalize(magnitude, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

# Direction: arctan(Gy/Gx)
direction = np.arctan2(grad_y, grad_x) * 180 / np.pi

fig, axes = plt.subplots(2, 3, figsize=(18, 11))
axes[0,0].imshow(gray, cmap='gray');           axes[0,0].set_title('Grayscale');       axes[0,0].axis('off')
axes[0,1].imshow(np.abs(grad_x), cmap='gray'); axes[0,1].set_title('Gradient X (vertical edges)');   axes[0,1].axis('off')
axes[0,2].imshow(np.abs(grad_y), cmap='gray'); axes[0,2].set_title('Gradient Y (horizontal edges)'); axes[0,2].axis('off')
axes[1,0].imshow(magnitude, cmap='gray');      axes[1,0].set_title('Gradient Magnitude (all edges)'); axes[1,0].axis('off')
axes[1,1].imshow(direction, cmap='hsv');       axes[1,1].set_title('Gradient Direction (angle = color)'); axes[1,1].axis('off')
axes[1,2].axis('off')
plt.suptitle('Gradient = rate of pixel change. High magnitude = strong edge.', fontsize=12)
plt.tight_layout(); plt.show()
```

## Key Takeaway
Gradient X detects vertical edges (left-right changes).
Gradient Y detects horizontal edges (top-bottom changes).
Magnitude combines both. Direction tells you which way the edge faces.
