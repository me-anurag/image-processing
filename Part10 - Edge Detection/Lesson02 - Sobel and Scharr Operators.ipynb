# Lesson 02 — Sobel and Scharr Operators

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Sobel
sobel_x  = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_y  = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_xy = cv2.magnitude(sobel_x, sobel_y)

# Scharr — more accurate than Sobel for rotational symmetry
scharr_x  = cv2.Scharr(gray, cv2.CV_64F, 1, 0)
scharr_y  = cv2.Scharr(gray, cv2.CV_64F, 0, 1)
scharr_xy = cv2.magnitude(scharr_x, scharr_y)

# Normalize for display
def norm(img):
    return cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

fig, axes = plt.subplots(2, 3, figsize=(18, 11))
for ax, im, t in zip(axes.flat,
    [norm(sobel_x), norm(sobel_y), norm(sobel_xy),
     norm(scharr_x), norm(scharr_y), norm(scharr_xy)],
    ['Sobel X', 'Sobel Y', 'Sobel Combined',
     'Scharr X', 'Scharr Y', 'Scharr Combined']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.suptitle('Scharr is more rotationally symmetric than Sobel', fontsize=12)
plt.tight_layout(); plt.show()

# ksize comparison for Sobel
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, ks in zip(axes, [1, 3, 5, 7]):
    sx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=ks)
    sy = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=ks)
    mag = norm(cv2.magnitude(sx, sy))
    ax.imshow(mag, cmap='gray'); ax.set_title(f'Sobel ksize={ks}'); ax.axis('off')
plt.suptitle('Larger ksize = smoother edges, less noise sensitivity', fontsize=12)
plt.show()
```

## Key Takeaway
Use Scharr when you need precise gradient direction (tracking, optical flow).
Use Sobel ksize=3 for general edge detection.
Always use CV_64F output dtype to preserve negative gradient values before computing magnitude.
