# Lesson 03 — Laplacian

## Why
Laplacian computes the second derivative — it detects edges in all directions at once.
It's also used to measure image sharpness (blur detection).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Laplacian
lap = cv2.Laplacian(gray, cv2.CV_64F, ksize=3)
lap_abs = cv2.convertScaleAbs(lap)

# Laplacian of Gaussian (LoG) — blur first to reduce noise sensitivity
blurred = cv2.GaussianBlur(gray, (5,5), 0)
log     = cv2.Laplacian(blurred, cv2.CV_64F, ksize=3)
log_abs = cv2.convertScaleAbs(log)

# Use Laplacian variance to measure sharpness
def measure_sharpness(img_gray):
    lap_var = cv2.Laplacian(img_gray, cv2.CV_64F).var()
    return lap_var

sharp_var  = measure_sharpness(gray)
blurry_img = cv2.GaussianBlur(gray, (15,15), 0)
blurry_var = measure_sharpness(blurry_img)

print(f"Sharp image Laplacian variance:  {sharp_var:.1f}")
print(f"Blurry image Laplacian variance: {blurry_var:.1f}")
print(f"Rule of thumb: variance < 100 → image is blurry")

fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, im, t in zip(axes,
    [gray, blurry_img, lap_abs, log_abs],
    ['Original', 'Blurred', 'Laplacian', 'LoG (Gaussian first)']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Laplacian variance is the standard blur detection metric.
Low variance = blurry. High variance = sharp.
Always apply Gaussian blur before Laplacian (LoG) to suppress noise amplification.
