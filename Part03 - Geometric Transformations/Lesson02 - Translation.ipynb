# Lesson 02 — Translation

## Why This Lesson
Translation moves an image in X or Y. It's the simplest affine operation and teaches you how `warpAffine` works — which handles all affine operations.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
h, w = img.shape[:2]

# Translation matrix: [[1, 0, tx], [0, 1, ty]]
tx, ty = 150, 80

M = np.float32([[1, 0, tx],
                [0, 1, ty]])

translated = cv2.warpAffine(img, M, (w, h))

# Negative translation (shift left and up)
M_neg = np.float32([[1, 0, -tx], [0, 1, -ty]])
translated_neg = cv2.warpAffine(img, M_neg, (w, h))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, im, t in zip(axes,
    [img, translated, translated_neg],
    ['Original', f'Right {tx}px, Down {ty}px', f'Left {tx}px, Up {ty}px']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
`warpAffine(img, M, (w, h))` — the output size stays the same. Translated-out pixels are filled with black.
