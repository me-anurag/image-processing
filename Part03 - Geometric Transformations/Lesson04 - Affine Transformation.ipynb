# Lesson 04 — Affine Transformation

## Why This Lesson
Affine transformations preserve parallel lines. They unify translation, rotation, scaling, and shear into a single 2×3 matrix. Any combination of these is one `warpAffine` call.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
h, w = img.shape[:2]

# Define 3 point correspondences: where 3 points go after transformation
src_pts = np.float32([[0,     0    ],
                       [w-1,  0    ],
                       [0,    h-1  ]])

# Shear effect: move points
dst_pts = np.float32([[w*0.1,  h*0.1],
                       [w*0.9,  h*0.05],
                       [w*0.05, h*0.9 ]])

M = cv2.getAffineTransform(src_pts, dst_pts)
warped = cv2.warpAffine(img, M, (w, h))

print("Affine Matrix:")
print(M)
print("\nAffine preserves: parallel lines, ratios of distances along lines")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, im, t in zip(axes, [img, warped], ['Original', 'Affine Warped']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()

# Manual affine matrix for pure rotation + scale
angle = 30 * np.pi / 180  # radians
scale = 0.8
M_manual = np.float32([
    [scale*np.cos(angle), -scale*np.sin(angle), w*(1-scale*np.cos(angle))/2],
    [scale*np.sin(angle),  scale*np.cos(angle), h*(1-scale*np.cos(angle))/2]
])
manual_result = cv2.warpAffine(img, M_manual, (w, h))
plt.imshow(cv2.cvtColor(manual_result, cv2.COLOR_BGR2RGB))
plt.title('Manual matrix: 30° rotation + 0.8x scale'); plt.axis('off'); plt.show()
```

## Key Takeaway
3 point pairs uniquely define any affine transformation. `warpAffine` applies it to every pixel at once.
