# Lesson 03 — Rotation and Flipping

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
h, w = img.shape[:2]
cx, cy = w//2, h//2

# Rotation: center, angle (CCW), scale
angles = [15, 45, 90, 180]
fig, axes = plt.subplots(1, 5, figsize=(25, 5))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); axes[0].set_title('Original'); axes[0].axis('off')

for ax, angle in zip(axes[1:], angles):
    M = cv2.getRotationMatrix2D((cx, cy), angle, 1.0)
    rotated = cv2.warpAffine(img, M, (w, h))
    ax.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
    ax.set_title(f'Rotate {angle}°'); ax.axis('off')
plt.suptitle('Rotation around center — black corners appear at non-90° angles')
plt.show()

# Flips
flip_h = cv2.flip(img, 1)   # 1  = horizontal mirror
flip_v = cv2.flip(img, 0)   # 0  = vertical mirror
flip_b = cv2.flip(img, -1)  # -1 = both axes

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, im, t in zip(axes, [img, flip_h, flip_v, flip_b],
    ['Original','Flip Horizontal','Flip Vertical','Flip Both']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
`getRotationMatrix2D((cx,cy), angle, scale)` — angle is counterclockwise degrees. Positive = CCW.
