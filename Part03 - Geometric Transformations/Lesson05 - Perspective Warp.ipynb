# Lesson 05 — Perspective Warp

## Why This Lesson
Perspective warp is how every document scanner works. It corrects the 3D perspective distortion that happens when a camera isn't perfectly flat above a subject.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
h, w = img.shape[:2]

# Simulate a perspective distortion first
# then unwarp it back
src = np.float32([
    [w*0.15, h*0.15],   # top-left — pushed inward
    [w*0.85, h*0.10],   # top-right
    [w*0.90, h*0.90],   # bottom-right
    [w*0.10, h*0.88]    # bottom-left
])

# Destination: perfect rectangle
dst = np.float32([
    [0,   0  ],
    [w-1, 0  ],
    [w-1, h-1],
    [0,   h-1]
])

M = cv2.getPerspectiveTransform(src, dst)
warped = cv2.warpPerspective(img, M, (w, h))

# Visualize source polygon
vis = img.copy()
pts = src.astype(np.int32).reshape((-1,1,2))
cv2.polylines(vis, [pts], True, (0,255,0), 3)
for pt in src:
    cv2.circle(vis, tuple(pt.astype(int)), 8, (0,0,255), -1)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); axes[0].set_title('Source region (green polygon)'); axes[0].axis('off')
axes[1].imshow(cv2.cvtColor(warped, cv2.COLOR_BGR2RGB)); axes[1].set_title('Perspective corrected'); axes[1].axis('off')
plt.show()

# Inverse warp: project flat rectangle into perspective
M_inv = cv2.getPerspectiveTransform(dst, src)
reprojected = cv2.warpPerspective(warped, M_inv, (w, h))
```

## Key Takeaway
4 point pairs → `getPerspectiveTransform` → `warpPerspective`. This is the math behind CamScanner, Google PhotoScan, and every document digitization tool.
