# Lesson 01 — Resize and Rescale

## Why This Lesson
Resizing is the most common operation in any CV pipeline. Wrong interpolation = blurry or blocky results. Knowing which method to use when is a real skill.

## Interpolation Methods
| Method | Best For | Speed |
|---|---|---|
| INTER_NEAREST | Binary masks | Fastest |
| INTER_LINEAR | Default, general use | Fast |
| INTER_CUBIC | Upscaling, high quality | Slow |
| INTER_AREA | Downscaling (no aliasing) | Medium |

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
h, w = img.shape[:2]

methods = [cv2.INTER_NEAREST, cv2.INTER_LINEAR, cv2.INTER_CUBIC, cv2.INTER_AREA]
names = ['NEAREST', 'LINEAR (default)', 'CUBIC', 'AREA']

# Downscale 4x then upscale back — shows quality difference clearly
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, method, name in zip(axes, methods, names):
    small = cv2.resize(img, (w//4, h//4), interpolation=method)
    back  = cv2.resize(small, (w, h), interpolation=method)
    ax.imshow(cv2.cvtColor(back, cv2.COLOR_BGR2RGB))
    ax.set_title(name); ax.axis('off')
plt.suptitle('Downscale 4x then upscale — quality difference is visible', fontsize=13)
plt.tight_layout(); plt.show()

# Resize to exact dimensions
resized_exact = cv2.resize(img, (640, 480))
print(f"Original: {img.shape[:2]} → Resized: {resized_exact.shape[:2]}")

# Resize by scale factor (keep aspect ratio)
scale = 0.5
resized_scale = cv2.resize(img, None, fx=scale, fy=scale, interpolation=cv2.INTER_AREA)
print(f"Half size: {resized_scale.shape[:2]}")
```

## Activity
Resize an image to 1/8 its size using all 4 methods. Display them in a 2×2 grid. Which looks the least blocky?

## Key Takeaway
Use `INTER_AREA` when shrinking. Use `INTER_CUBIC` when enlarging. `INTER_NEAREST` only for binary masks.
