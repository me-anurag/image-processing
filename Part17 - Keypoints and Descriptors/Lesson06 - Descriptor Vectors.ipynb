# Lesson 06 — Descriptor Vectors: What They Encode

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sift = cv2.SIFT_create(nfeatures=5)
kps, descs = sift.detectAndCompute(gray, None)

print("Each SIFT descriptor encodes:")
print("  - Gradient orientations in 4x4 spatial bins")
print("  - 8 orientation bins per spatial bin")
print("  - 4x4x8 = 128 values total")
print(f"\nDescriptor for keypoint 0: shape={descs[0].shape}")
print(f"Min={descs[0].min():.0f}  Max={descs[0].max():.0f}  Mean={descs[0].mean():.1f}")

# Visualize descriptor as an 8x16 grid
fig, axes = plt.subplots(1, min(5,len(kps)), figsize=(20, 4))
for ax, kp, desc in zip(axes, kps[:5], descs[:5]):
    ax.imshow(desc.reshape(8,16), cmap='hot', aspect='auto')
    ax.set_title(f'KP at ({int(kp.pt[0])},{int(kp.pt[1])})')
    ax.axis('off')
plt.suptitle('SIFT descriptors visualized — each column = one spatial bin (8 orientations)', fontsize=11)
plt.tight_layout(); plt.show()
```

## Key Takeaway
A descriptor is just a vector of numbers. Two descriptors are "similar" if their Euclidean (SIFT)
or Hamming (ORB) distance is small. This distance is what feature matching computes.
