# Lesson 03 — Shi-Tomasi: Good Features to Track

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Shi-Tomasi: returns actual corner POINTS (more convenient than Harris response map)
corners = cv2.goodFeaturesToTrack(gray, maxCorners=200, qualityLevel=0.01,
                                   minDistance=10, blockSize=3)

vis = img.copy()
if corners is not None:
    corners = np.int0(corners)
    for pt in corners:
        x, y = pt.ravel()
        cv2.circle(vis, (x,y), 4, (0,255,0), -1)
    print(f"Found {len(corners)} corners")

plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title(f'Shi-Tomasi: {len(corners) if corners is not None else 0} corners'); plt.axis('off'); plt.show()
```

## Key Takeaway
Shi-Tomasi is the improved Harris. Use `goodFeaturesToTrack` when you need corner POINTS (e.g. for Lucas-Kanade tracking).
It's faster and more stable than Harris for tracking applications.
