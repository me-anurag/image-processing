# Lesson 05 — ORB: Fast, Free, Good Enough

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# ORB vs SIFT: speed comparison
sift = cv2.SIFT_create()
orb  = cv2.ORB_create(nfeatures=500)

t0 = time.time()
kps_sift, desc_sift = sift.detectAndCompute(gray, None)
t_sift = time.time() - t0

t0 = time.time()
kps_orb, desc_orb = orb.detectAndCompute(gray, None)
t_orb = time.time() - t0

print(f"SIFT: {len(kps_sift)} kps in {t_sift*1000:.1f}ms | 128-dim float descriptors")
print(f"ORB:  {len(kps_orb)}  kps in {t_orb*1000:.1f}ms  | 32-byte binary descriptors")
print(f"ORB is ~{t_sift/t_orb:.0f}x faster than SIFT")

vis_sift = cv2.drawKeypoints(img, kps_sift, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
vis_orb  = cv2.drawKeypoints(img, kps_orb,  None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

fig, axes = plt.subplots(1,2,figsize=(14,5))
axes[0].imshow(cv2.cvtColor(vis_sift, cv2.COLOR_BGR2RGB)); axes[0].set_title(f'SIFT ({len(kps_sift)} kps)'); axes[0].axis('off')
axes[1].imshow(cv2.cvtColor(vis_orb,  cv2.COLOR_BGR2RGB)); axes[1].set_title(f'ORB  ({len(kps_orb)} kps)');  axes[1].axis('off')
plt.show()
```

## Key Takeaway
Use ORB for real-time applications (fast, binary descriptors, free license).
Use SIFT when accuracy matters more than speed (research, offline matching).
