# Lesson 02 — Harris Corner Detector

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)

# Harris corner response
harris = cv2.cornerHarris(gray, blockSize=2, ksize=3, k=0.04)
harris = cv2.dilate(harris, None)  # slightly enlarge for visibility

# Mark corners where response is strong
vis = img.copy()
vis[harris > 0.01 * harris.max()] = [0,0,255]

# Also try on rotated image — Harris is rotation invariant
M      = cv2.getRotationMatrix2D((img.shape[1]//2, img.shape[0]//2), 30, 1)
rotated = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))
gray_r  = cv2.cvtColor(rotated, cv2.COLOR_BGR2GRAY).astype(np.float32)
harris_r = cv2.cornerHarris(gray_r, 2, 3, 0.04)
vis_r    = rotated.copy()
vis_r[harris_r > 0.01 * harris_r.max()] = [0,0,255]

fig, axes = plt.subplots(1, 3, figsize=(18,5))
for ax, im, t in zip(axes, [img, vis, vis_r],
    ['Original', 'Harris Corners (red)', 'Rotated — same corners found']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Harris responds to corners (high gradient change in both X and Y).
It is rotation-invariant but NOT scale-invariant — it misses corners at different zoom levels.
