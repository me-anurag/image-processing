# Lesson 04 — SIFT: Scale-Invariant Feature Transform

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sift = cv2.SIFT_create(nfeatures=500)
kps, descs = sift.detectAndCompute(gray, None)

print(f"SIFT keypoints: {len(kps)}")
print(f"Descriptor shape: {descs.shape}  (128-dimensional float vectors)")

# Test scale invariance: find keypoints at half size
small = cv2.resize(gray, None, fx=0.5, fy=0.5)
kps_s, descs_s = sift.detectAndCompute(small, None)

vis = cv2.drawKeypoints(img, kps, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title(f'SIFT: {len(kps)} keypoints — circles show detected scale'); plt.axis('off'); plt.show()
print(f"\nFull image: {len(kps)} keypoints.  Half size: {len(kps_s)} keypoints.")
print("SIFT finds keypoints at multiple scales — that's the S in SIFT.")
```

## Key Takeaway
SIFT is scale AND rotation invariant. It builds a 128-dim descriptor using gradient histograms.
It's patented (free since 2020). Still the gold standard for matching across scale changes.
