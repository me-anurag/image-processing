# Lesson 01 — What is a Keypoint

## Why This Lesson
A keypoint is a special point in an image: repeatable, distinctive, findable again under different conditions.
A descriptor is a vector of numbers that describes the patch around that keypoint — its "fingerprint".

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect ORB keypoints (fast overview)
orb = cv2.ORB_create(nfeatures=500)
kps, descs = orb.detectAndCompute(gray, None)

print(f"Found {len(kps)} keypoints")
print(f"Descriptor shape: {descs.shape}  (each row = 1 keypoint fingerprint)")
print(f"Descriptor dtype: {descs.dtype}  (ORB uses binary descriptors)")

# Draw keypoints — size shows scale, orientation shows angle
vis = cv2.drawKeypoints(img, kps, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

# Show a single descriptor vector
print(f"\nDescriptor of first keypoint (32 bytes = 256 bits):\n{descs[0]}")

plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title(f'{len(kps)} ORB keypoints — circle size=scale, line=orientation')
plt.axis('off'); plt.show()
```

## Key Takeaway
Keypoint = location + scale + orientation. Descriptor = 128 or 256 numbers describing what the patch looks like.
Same physical point in two photos → similar descriptor → can be matched.
