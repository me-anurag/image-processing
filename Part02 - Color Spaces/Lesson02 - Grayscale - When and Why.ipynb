# Lesson 02 — Grayscale: When and Why

## Why This Lesson
Most image processing algorithms work on single-channel images. Grayscale removes color and keeps only luminance — the brightness information. Knowing when to go grayscale is a practical skill.

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Method 1: OpenCV conversion
gray_cv = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Method 2: Manual average (naive — not how OpenCV does it)
gray_manual = img.mean(axis=2).astype(np.uint8)

# Method 3: Weighted (human perception — how OpenCV actually does it)
# 0.114*B + 0.587*G + 0.299*R
b, g, r = cv2.split(img)
gray_weighted = (0.114*b + 0.587*g + 0.299*r).astype(np.uint8)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, im, t in zip(axes,
    [gray_cv, gray_manual, gray_weighted],
    ['OpenCV Grayscale', 'Simple Average', 'Weighted (perceptual)']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.show()

print("Difference between CV and weighted:", np.abs(gray_cv.astype(int) - gray_weighted.astype(int)).max())
```

## Key Takeaway
Green contributes most to perceived brightness (0.587). Blue contributes least (0.114). This is why `cv2.cvtColor` gives different results than a simple average — it models human vision.
