# Lesson 04 — LAB Color Space: Perceptual Uniformity

## Why This Lesson
LAB is designed so that equal numerical distances mean equal perceptual differences. This makes it ideal for color comparison, skin detection, and any task where "how different do these colors look" matters.

## Concepts
- **L**: Lightness (0=black, 100=white)
- **A**: Green↔Red axis
- **B**: Blue↔Yellow axis

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

L, A, B = cv2.split(lab)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, im, t in zip(axes,
    [cv2.cvtColor(img, cv2.COLOR_BGR2RGB), L, A, B],
    ['Original', 'L (Lightness)', 'A (Green-Red)', 'B (Blue-Yellow)']):
    ax.imshow(im if t == 'Original' else im, cmap='gray' if t != 'Original' else None)
    ax.set_title(t); ax.axis('off')
plt.show()

# Practical use: compare color similarity
color1 = np.uint8([[[100, 150, 200]]])   # BGR
color2 = np.uint8([[[110, 145, 195]]])   # similar BGR

lab1 = cv2.cvtColor(color1, cv2.COLOR_BGR2LAB).astype(float)
lab2 = cv2.cvtColor(color2, cv2.COLOR_BGR2LAB).astype(float)

delta_e = np.sqrt(np.sum((lab1 - lab2)**2))
print(f"Color difference (Delta-E): {delta_e:.2f}")
print("< 2.3 means barely perceptible difference")
```

## Key Takeaway
Use LAB when you need to measure or compare colors perceptually. Use HSV when you need to select colors by hue range. They solve different problems.
