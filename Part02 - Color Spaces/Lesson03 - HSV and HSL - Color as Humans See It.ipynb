# Lesson 03 — HSV and HSL: Color as Humans See It

## Why This Lesson
RGB is how screens work. HSV is how humans think about color. When you want to isolate "the red object" or "the dark region," HSV makes it trivial. In RGB it's a nightmare.

## Concepts
- **H (Hue)**: the color itself — 0°=Red, 60°=Yellow, 120°=Green, 180°=Cyan, 240°=Blue, 300°=Magenta
- **S (Saturation)**: how pure/vivid the color is — 0=grey, 255=full color
- **V (Value)**: brightness — 0=black, 255=bright

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(hsv)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
imgs = [cv2.cvtColor(img, cv2.COLOR_BGR2RGB), h, s, v]
titles = ['Original', 'Hue', 'Saturation', 'Value']
cmaps = [None, 'hsv', 'gray', 'gray']

for ax, im, t, c in zip(axes, imgs, titles, cmaps):
    ax.imshow(im, cmap=c); ax.set_title(t); ax.axis('off')
plt.show()

# Isolate RED objects using HSV range
lower_red1 = np.array([0, 120, 70])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])

mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
mask2 = cv2.inRange(hsv, lower_red2, upper_red2)
red_mask = cv2.bitwise_or(mask1, mask2)

result = cv2.bitwise_and(img, img, mask=red_mask)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title('Only Red Objects'); plt.axis('off')
plt.show()
```

## Key Takeaway
In OpenCV, H ranges 0–179 (not 0–360). S and V range 0–255. Red wraps around (near 0 AND near 179), so always use two ranges for red.
