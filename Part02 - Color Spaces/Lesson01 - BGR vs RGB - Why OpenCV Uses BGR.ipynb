# Lesson 01 — BGR vs RGB: Why OpenCV Uses BGR

## Why This Lesson
OpenCV loads images in BGR order, not RGB. Every beginner gets burned by this once. After this lesson, never again.

## The Confusion
Most tools (matplotlib, PIL, cameras) use RGB. OpenCV uses BGR — Blue, Green, Red. Same three channels, opposite order. If you display a BGR image in matplotlib without converting, red becomes blue.

## Code
```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img_bgr = cv2.imread('sample.jpg')   # OpenCV loads as BGR

# WRONG — colors are swapped
plt.subplot(1,2,1)
plt.imshow(img_bgr)
plt.title('Wrong (BGR shown as RGB)')

# CORRECT — convert first
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.subplot(1,2,2)
plt.imshow(img_rgb)
plt.title('Correct (BGR→RGB)')
plt.show()

# Proof: split and compare
b, g, r = cv2.split(img_bgr)
print("img[:,:,0] is Blue:", np.array_equal(b, img_bgr[:,:,0]))
print("img[:,:,2] is Red:", np.array_equal(r, img_bgr[:,:,2]))
```

## Key Takeaway
OpenCV = BGR. Matplotlib = RGB. Always convert with `cv2.COLOR_BGR2RGB` before `plt.imshow`. Or use `img[:,:,::-1]` as a quick flip.
