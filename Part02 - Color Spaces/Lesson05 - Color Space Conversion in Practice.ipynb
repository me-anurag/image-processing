# Lesson 05 — Color Space Conversion in Practice

## Why This Lesson
Bringing all color spaces together. Knowing which space to use for which problem is a practical skill you'll apply in every pipeline.

## The Decision Framework
```
Need to display?          → RGB (matplotlib) or BGR (cv2.imshow)
Need to isolate by color? → HSV
Need to separate light from color? → LAB or HSV (use V/L channel)
Need to denoise? → LAB (denoise only L channel, not color channels)
Need to compare colors perceptually? → LAB (Delta-E distance)
```

## Code
```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

conversions = {
    'BGR (original)': img,
    'RGB': cv2.cvtColor(img, cv2.COLOR_BGR2RGB),
    'Grayscale': cv2.cvtColor(img, cv2.COLOR_BGR2GRAY),
    'HSV': cv2.cvtColor(img, cv2.COLOR_BGR2HSV),
    'LAB': cv2.cvtColor(img, cv2.COLOR_BGR2LAB),
    'YCrCb': cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb),
}

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, (name, im) in zip(axes.flat, conversions.items()):
    if len(im.shape) == 2:
        ax.imshow(im, cmap='gray')
    else:
        ax.imshow(im)
    ax.set_title(name); ax.axis('off')
plt.tight_layout(); plt.show()

# Practical: change sky color by manipulating HSV H channel
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
# Shift all blues (sky) toward cyan
sky_mask = cv2.inRange(hsv, np.array([100,50,50]), np.array([130,255,255]))
hsv[:,:,0][sky_mask > 0] = 90   # shift hue to cyan
result = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
```

## Key Takeaway
Color space choice is a tool selection problem. There is no single best space — only the best space for your current task.
