# Lesson 04 — Visualizing Optical Flow

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def flow_to_rgb(flow):
    mag, ang = cv2.cartToPolar(flow[...,0], flow[...,1])
    hsv = np.zeros((flow.shape[0],flow.shape[1],3), dtype=np.uint8)
    hsv[...,0] = ang * 90 / np.pi
    hsv[...,1] = 255
    hsv[...,2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)
    return cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

cap = cv2.VideoCapture('sample_video.mp4')
ret, prev = cap.read()
pg = cv2.cvtColor(prev, cv2.COLOR_BGR2GRAY)

flows = []
for _ in range(10):
    ret, curr = cap.read()
    if not ret: break
    cg = cv2.cvtColor(curr, cv2.COLOR_BGR2GRAY)
    flows.append(flow_to_rgb(cv2.calcOpticalFlowFarneback(pg,cg,None,0.5,3,15,3,5,1.2,0)))
    pg = cg.copy()
cap.release()

if flows:
    fig, axes = plt.subplots(2,5,figsize=(25,9))
    for ax, fl in zip(axes.flat, flows[:10]):
        ax.imshow(fl); ax.axis('off')
    plt.suptitle('Dense optical flow across 10 consecutive frames',fontsize=12)
    plt.tight_layout(); plt.show()
```

## Key Takeaway
Color wheel: right=red, up=cyan, left=teal, down=yellow/green. Bright=fast, dark=slow/still.
