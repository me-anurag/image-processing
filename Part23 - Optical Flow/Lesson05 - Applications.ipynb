# Lesson 05 — Optical Flow Applications

```python
import cv2
import numpy as np

print("OPTICAL FLOW APPLICATIONS:")
print()
print("1. VIDEO STABILIZATION")
print("   Compute flow between frames → estimate global camera motion → warp to remove shake")
print()
print("2. ACTION RECOGNITION")
print("   Build 'motion history image' from flow magnitude → classify action type")
print()
print("3. AUTONOMOUS DRIVING")
print("   Objects expanding in flow field = approaching. Contracting = moving away.")
print()
print("4. GESTURE RECOGNITION")
print("   Track hand flow direction → left/right/up/down gesture")
print()
print("5. OBJECT SEGMENTATION")
print("   Moving objects have high flow magnitude → threshold flow to segment them")

# Demo: motion segmentation from flow
cap = cv2.VideoCapture('sample_video.mp4')
ret, f1 = cap.read()
ret, f2 = cap.read()
cap.release()

if ret:
    g1 = cv2.cvtColor(f1, cv2.COLOR_BGR2GRAY)
    g2 = cv2.cvtColor(f2, cv2.COLOR_BGR2GRAY)
    flow = cv2.calcOpticalFlowFarneback(g1,g2,None,0.5,3,15,3,5,1.2,0)
    mag, _ = cv2.cartToPolar(flow[...,0], flow[...,1])
    motion_mask = (mag > 1.0).astype(np.uint8) * 255
    import matplotlib.pyplot as plt
    fig,axes=plt.subplots(1,2,figsize=(14,5))
    axes[0].imshow(cv2.cvtColor(f2,cv2.COLOR_BGR2RGB)); axes[0].set_title('Frame'); axes[0].axis('off')
    axes[1].imshow(motion_mask,cmap='gray'); axes[1].set_title('Motion segmentation from flow'); axes[1].axis('off')
    plt.show()
```

## Key Takeaway
Optical flow is a mid-level feature — richer than edges, more interpretable than raw pixels.
It's the foundation of video understanding.
