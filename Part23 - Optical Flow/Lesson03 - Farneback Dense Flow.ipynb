# Lesson 03 — Farneback Dense Optical Flow

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')
ret, f1 = cap.read()
for _ in range(5): ret, f2 = cap.read()
cap.release()

g1 = cv2.cvtColor(f1, cv2.COLOR_BGR2GRAY)
g2 = cv2.cvtColor(f2, cv2.COLOR_BGR2GRAY)

flow = cv2.calcOpticalFlowFarneback(
    g1, g2, None,
    pyr_scale=0.5,   # image pyramid scale
    levels=3,         # pyramid levels
    winsize=15,       # averaging window
    iterations=3,
    poly_n=5,         # neighborhood size
    poly_sigma=1.2,   # Gaussian sigma
    flags=0
)

# Arrow visualization
step = 20
y,x  = np.mgrid[step//2:g1.shape[0]:step, step//2:g1.shape[1]:step]
fx   = flow[y,x,0]
fy   = flow[y,x,1]

fig, axes = plt.subplots(1,2,figsize=(16,6))
mag, ang = cv2.cartToPolar(flow[...,0], flow[...,1])
hsv = np.zeros((*g1.shape,3),dtype=np.uint8)
hsv[...,0]=ang*90/np.pi; hsv[...,1]=255
hsv[...,2]=cv2.normalize(mag,None,0,255,cv2.NORM_MINMAX)
axes[0].imshow(cv2.cvtColor(hsv,cv2.COLOR_HSV2RGB)); axes[0].set_title('Dense flow (color)'); axes[0].axis('off')
axes[1].imshow(g1,cmap='gray')
axes[1].quiver(x,y,fx,fy,color='yellow',scale=50,width=0.003)
axes[1].set_title('Arrow field'); axes[1].axis('off')
plt.show()
```

## Key Takeaway
Farneback gives flow for EVERY pixel. Computationally heavier than Lucas-Kanade but gives full motion field.
