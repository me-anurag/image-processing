# Lesson 01 — What is Optical Flow

## Why
Optical flow measures the apparent motion of every pixel between two frames.
It assumes brightness constancy: a pixel at (x,y) in frame t has the same brightness at (x+dx,y+dy) in frame t+1.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')
ret, f1 = cap.read()
ret, f2 = cap.read()
cap.release()

g1 = cv2.cvtColor(f1, cv2.COLOR_BGR2GRAY)
g2 = cv2.cvtColor(f2, cv2.COLOR_BGR2GRAY)

# Dense optical flow: every pixel gets a flow vector
flow = cv2.calcOpticalFlowFarneback(g1, g2, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# Visualize as HSV: angle=hue, magnitude=saturation
mag, ang = cv2.cartToPolar(flow[...,0], flow[...,1])
hsv = np.zeros((*g1.shape, 3), dtype=np.uint8)
hsv[...,0] = ang * 90 / np.pi       # hue = direction
hsv[...,1] = 255                     # full saturation
hsv[...,2] = cv2.normalize(mag,None,0,255,cv2.NORM_MINMAX)
rgb_flow = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

fig, axes = plt.subplots(1,3,figsize=(18,5))
axes[0].imshow(cv2.cvtColor(f1,cv2.COLOR_BGR2RGB)); axes[0].set_title('Frame 1'); axes[0].axis('off')
axes[1].imshow(cv2.cvtColor(f2,cv2.COLOR_BGR2RGB)); axes[1].set_title('Frame 2'); axes[1].axis('off')
axes[2].imshow(rgb_flow);                            axes[2].set_title('Optical flow: color=direction, brightness=speed'); axes[2].axis('off')
plt.show()
```

## Key Takeaway
Color encodes direction of motion. Brightness encodes speed.
Static background = dark (no motion). Moving objects = bright and colored.
