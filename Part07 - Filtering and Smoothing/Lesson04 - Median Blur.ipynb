# Lesson 04 — Median Blur

## Why This Lesson
Median blur replaces each pixel with the median of its neighbors — not the average.
This makes it immune to extreme outliers like salt-and-pepper noise, which Gaussian cannot handle.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Add salt-and-pepper noise manually
def add_salt_pepper(img, amount=0.05):
    noisy = img.copy()
    total = int(amount * img.size)
    # Salt (white dots)
    coords = [np.random.randint(0, i, total//2) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = 255
    # Pepper (black dots)
    coords = [np.random.randint(0, i, total//2) for i in img.shape[:2]]
    noisy[coords[0], coords[1]] = 0
    return noisy

noisy = add_salt_pepper(img, 0.04)

gaussian = cv2.GaussianBlur(noisy, (5,5), 0)
median3  = cv2.medianBlur(noisy, 3)
median5  = cv2.medianBlur(noisy, 5)

fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, im, t in zip(axes,
    [noisy, gaussian, median3, median5],
    ['Salt & Pepper noise', 'Gaussian (smears dots)', 'Median k=3 (clean!)', 'Median k=5 (cleaner)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.suptitle('Median blur destroys salt-and-pepper noise. Gaussian cannot.', fontsize=12)
plt.show()
```

## Key Takeaway
Median blur for salt-and-pepper noise. Gaussian for Gaussian noise. Know which noise type you have
before choosing a filter — wrong filter = worse result.
