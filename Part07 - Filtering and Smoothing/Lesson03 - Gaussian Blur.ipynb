# Lesson 03 — Gaussian Blur

## Why This Lesson
Gaussian blur is the most important blur in all of image processing. Every edge detector,
every CNN preprocessing step, every camera uses it internally. Sigma is everything.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Gaussian blur: weights follow bell curve — center pixel matters most
sigmas = [0.5, 1, 2, 5, 15]
fig, axes = plt.subplots(1, 6, figsize=(26, 5))
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); axes[0].set_title('Original'); axes[0].axis('off')

for ax, s in zip(axes[1:], sigmas):
    blurred = cv2.GaussianBlur(img, (0,0), s)  # ksize=(0,0) → auto from sigma
    ax.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
    ax.set_title(f'σ={s}'); ax.axis('off')
plt.suptitle('Gaussian blur: sigma controls the spread of the bell curve', fontsize=12)
plt.tight_layout(); plt.show()

# Add Gaussian noise then remove
noisy = img.copy().astype(np.float32)
noisy += np.random.normal(0, 30, img.shape)
noisy  = np.clip(noisy, 0, 255).astype(np.uint8)

denoised = cv2.GaussianBlur(noisy, (0,0), 2)

fig, axes = plt.subplots(1,3,figsize=(18,5))
for ax, im, t in zip(axes, [img, noisy, denoised], ['Original','Noisy (σ=30)','Gaussian denoised (σ=2)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Gaussian blur = low-pass filter. It removes high-frequency content (noise, edges).
Sigma controls how aggressively. σ=1 is gentle, σ=5 destroys all detail.
