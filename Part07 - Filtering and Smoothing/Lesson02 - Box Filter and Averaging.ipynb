# Lesson 02 — Box Filter and Averaging

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Add noise to demonstrate filtering
noisy = img.copy().astype(np.float32)
noisy += np.random.normal(0, 25, img.shape)
noisy  = np.clip(noisy, 0, 255).astype(np.uint8)

# Box filter (uniform average)
sizes = [3, 7, 15, 31]
fig, axes = plt.subplots(1, 5, figsize=(25, 5))
axes[0].imshow(cv2.cvtColor(noisy, cv2.COLOR_BGR2RGB)); axes[0].set_title('Noisy input'); axes[0].axis('off')

for ax, k in zip(axes[1:], sizes):
    blurred = cv2.blur(noisy, (k, k))
    ax.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
    ax.set_title(f'Box blur {k}×{k}'); ax.axis('off')
plt.suptitle('Larger kernel = more blur, more noise removed, more edge loss', fontsize=12)
plt.show()
```

## Key Takeaway
Box blur is the simplest filter — every neighbor gets equal weight.
It removes noise but also smears edges. Use Gaussian or bilateral for better results.
