# Lesson 05 — Bilateral Filter: Edge-Preserving Smoothing

## Why This Lesson
Gaussian blur destroys edges. Bilateral filter blurs flat regions while preserving sharp edges.
It considers both spatial distance AND color similarity — pixels far in color don't get averaged together.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')

# Parameters: d=diameter, sigmaColor=color range, sigmaSpace=spatial range
gaussian  = cv2.GaussianBlur(img, (15,15), 0)
bilateral = cv2.bilateralFilter(img, d=15, sigmaColor=75, sigmaSpace=75)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
for ax, im, t in zip(axes,
    [img, gaussian, bilateral],
    ['Original', 'Gaussian (edges blurred)', 'Bilateral (edges preserved)']):
    ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); ax.set_title(t); ax.axis('off')
plt.suptitle('Bilateral filter: smooth skin, keep hair/eye edges sharp', fontsize=12)
plt.show()

# sigmaColor exploration
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, sc in zip(axes, [10, 30, 75, 150]):
    result = cv2.bilateralFilter(img, 15, sc, 75)
    ax.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    ax.set_title(f'sigmaColor={sc}'); ax.axis('off')
plt.suptitle('Higher sigmaColor = more colors averaged together = less edge preservation', fontsize=12)
plt.show()
```

## Key Takeaway
Bilateral filter is slow (O(n²) per pixel) but produces the best quality smoothing.
sigmaColor controls how different colors can be before they stop being averaged.
Low sigmaColor = hard edges preserved. High sigmaColor = approaches Gaussian blur.
