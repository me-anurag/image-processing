# Lesson 01 — inRange Masking

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('sample.jpg')
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Green mask (plants, leaves)
lower_green = np.array([35,  40, 40])
upper_green = np.array([85, 255,255])
mask_green  = cv2.inRange(hsv, lower_green, upper_green)

# Blue mask (sky, water)
lower_blue  = np.array([100, 50, 50])
upper_blue  = np.array([130,255,255])
mask_blue   = cv2.inRange(hsv, lower_blue, upper_blue)

result_green = cv2.bitwise_and(img, img, mask=mask_green)
result_blue  = cv2.bitwise_and(img, img, mask=mask_blue)

fig, axes = plt.subplots(1, 5, figsize=(26, 5))
for ax, im, t in zip(axes,
    [img, mask_green, result_green, mask_blue, result_blue],
    ['Original','Green mask','Green isolated','Blue mask','Blue isolated']):
    if len(im.shape)==2:
        ax.imshow(im, cmap='gray')
    else:
        ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
    ax.set_title(t); ax.axis('off')
plt.show()
```

## Key Takeaway
Always work in HSV for color masking. The H channel isolates color regardless of brightness.
