# Lesson 04 — Learning Rate: How Fast the Background Adapts

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')
frames = []
for _ in range(60):
    ret, f = cap.read()
    if ret: frames.append(f)
cap.release()

if frames:
    learning_rates = [-1, 0.001, 0.01, 0.1, 1.0]
    # -1 = auto (default ~0.005)
    results = {}
    for lr in learning_rates:
        mog = cv2.createBackgroundSubtractorMOG2(history=200, varThreshold=50)
        mask = None
        for f in frames:
            mask = mog.apply(f, learningRate=lr)
        results[lr] = mask

    fig, axes = plt.subplots(1, len(learning_rates), figsize=(26,5))
    for ax, (lr, mask) in zip(axes, results.items()):
        ax.imshow(mask, cmap='gray')
        ax.set_title(f'lr={lr}\n{"-1=auto" if lr==-1 else ""}')
        ax.axis('off')
    plt.suptitle('Learning rate: how fast background model updates\n'
                 'Low=stable but slow to adapt. High=adapts fast but foreground leaks in.',fontsize=11)
    plt.show()

print("lr = -1    : automatic (recommended for most cases)")
print("lr = 0     : background NEVER updates (freeze)")
print("lr = 1.0   : background updates to current frame every step (foreground disappears)")
print("lr = 0.005 : good balance — adapts in ~200 frames")
