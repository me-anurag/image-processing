# Lesson 03 — KNN Background Subtractor

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')

mog2 = cv2.createBackgroundSubtractorMOG2(history=200, varThreshold=50)
knn  = cv2.createBackgroundSubtractorKNN( history=200, dist2Threshold=400)

results_mog, results_knn = [], []
for i in range(50):
    ret, frame = cap.read()
    if not ret: break
    results_mog.append(mog2.apply(frame))
    results_knn.append(knn.apply(frame))

cap.release()

# Compare at frame 49
fig, axes = plt.subplots(1,2,figsize=(14,5))
axes[0].imshow(results_mog[-1], cmap='gray'); axes[0].set_title('MOG2 mask'); axes[0].axis('off')
axes[1].imshow(results_knn[-1], cmap='gray'); axes[1].set_title('KNN mask');  axes[1].axis('off')
plt.suptitle('MOG2 vs KNN — KNN is better for scenes with rapid illumination changes',fontsize=12)
plt.show()

print("MOG2: Gaussian mixture model — faster, standard choice")
print("KNN:  K-nearest neighbors    — better with sudden lighting changes")
