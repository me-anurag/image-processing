# Lesson 01 — Frame Differencing: The Naive Approach

## Why This Lesson
Frame differencing is the simplest possible motion detector.
Subtract two consecutive frames — wherever pixels changed, motion happened.
Naïve but fast, and understanding its limits motivates MOG2.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')

ret, prev = cap.read()
prev_gray = cv2.cvtColor(prev, cv2.COLOR_BGR2GRAY)

frames_shown = []
for _ in range(5):
    ret, curr = cap.read()
    if not ret: break
    curr_gray = cv2.cvtColor(curr, cv2.COLOR_BGR2GRAY)

    diff   = cv2.absdiff(prev_gray, curr_gray)
    _, motion = cv2.threshold(diff, 25, 255, cv2.THRESH_BINARY)

    frames_shown.append((curr_gray, diff, motion))
    prev_gray = curr_gray.copy()

cap.release()

# Show first result
if frames_shown:
    f, d, m = frames_shown[0]
    fig, axes = plt.subplots(1,3,figsize=(18,5))
    axes[0].imshow(f, cmap='gray');  axes[0].set_title('Current frame');     axes[0].axis('off')
    axes[1].imshow(d, cmap='gray');  axes[1].set_title('Absolute diff');      axes[1].axis('off')
    axes[2].imshow(m, cmap='gray');  axes[2].set_title('Motion mask (thresh)');axes[2].axis('off')
    plt.suptitle('Frame differencing: pixels that changed between frames',fontsize=12)
    plt.show()

print("Limitations of frame differencing:")
print("  - Camera shake = false motion everywhere")
print("  - Gradual lighting change = false motion")
print("  - Moving shadows counted as motion")
print("  - Stopped object disappears from mask immediately")
