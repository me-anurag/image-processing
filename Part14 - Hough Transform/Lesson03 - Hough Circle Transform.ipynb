# Lesson 03 — Hough Circle Transform

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blur = cv2.medianBlur(gray, 5)  # blur reduces false circles

circles = cv2.HoughCircles(
    blur,
    cv2.HOUGH_GRADIENT,
    dp=1,           # inverse ratio of accumulator resolution
    minDist=30,     # minimum distance between centers
    param1=50,      # Canny high threshold
    param2=30,      # accumulator threshold (lower = more circles)
    minRadius=10,
    maxRadius=100
)

vis = img.copy()
if circles is not None:
    circles = np.uint16(np.around(circles))
    for c in circles[0]:
        cv2.circle(vis, (c[0],c[1]), c[2], (0,255,0), 2)  # circle
        cv2.circle(vis, (c[0],c[1]), 2,    (0,0,255), 3)   # center
    print(f"Found {len(circles[0])} circles")

plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title('Original')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); plt.title('Detected Circles')
plt.show()
```

## Key Takeaway
Always median blur before HoughCircles — it's very sensitive to noise.
param2 is the most important knob: lower = more (but false) circles, higher = fewer (but accurate) circles.
