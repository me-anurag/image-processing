# Lesson 01 — Hough Line Transform

## Why This Lesson
Hough Transform detects straight lines even when they're partially hidden, noisy, or broken.
It works by voting in parameter space — elegant mathematics with real practical power.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 50, 150, apertureSize=3)

# Standard Hough: returns (rho, theta) pairs
lines = cv2.HoughLines(edges, rho=1, theta=np.pi/180, threshold=150)

vis = img.copy()
if lines is not None:
    for line in lines[:20]:
        rho, theta = line[0]
        a, b = np.cos(theta), np.sin(theta)
        x0, y0 = a*rho, b*rho
        x1 = int(x0 + 1000*(-b)); y1 = int(y0 + 1000*a)
        x2 = int(x0 - 1000*(-b)); y2 = int(y0 - 1000*a)
        cv2.line(vis, (x1,y1), (x2,y2), (0,0,255), 2)
    print(f"Found {len(lines)} lines")

plt.figure(figsize=(14,5))
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img,  cv2.COLOR_BGR2RGB)); plt.title('Original'); plt.axis('off')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(vis,  cv2.COLOR_BGR2RGB)); plt.title('Hough Lines'); plt.axis('off')
plt.show()
```

## Key Takeaway
Higher threshold = fewer, stronger lines. Lower = more lines including weak ones.
rho=1 (pixel), theta=π/180 (1 degree) are standard settings.
