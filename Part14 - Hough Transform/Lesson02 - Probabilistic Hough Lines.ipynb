# Lesson 02 — Probabilistic Hough Lines

## Why
Standard Hough returns full infinite lines. Probabilistic Hough returns actual line segments with start and end points — much more useful for real applications.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img   = cv2.imread('sample.jpg')
gray  = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 50, 150)

# Probabilistic Hough: returns (x1,y1,x2,y2) line segments
lines_p = cv2.HoughLinesP(edges, rho=1, theta=np.pi/180,
                           threshold=80, minLineLength=50, maxLineGap=10)

vis = img.copy()
if lines_p is not None:
    for line in lines_p:
        x1,y1,x2,y2 = line[0]
        cv2.line(vis, (x1,y1), (x2,y2), (0,255,0), 2)
    print(f"Found {len(lines_p)} line segments")

plt.figure(figsize=(14,5))
plt.subplot(1,2,1); plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); plt.title('Original')
plt.subplot(1,2,2); plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB)); plt.title('Probabilistic Hough Lines')
plt.show()
```

## Key Takeaway
Probabilistic Hough is almost always better than standard Hough for practical applications.
minLineLength filters short segments. maxLineGap connects nearby collinear segments.
