# Lesson 05 — When Tracking Fails: Occlusion, Fast Motion, ID Switching

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Demonstrate failure modes visually

# Failure 1: Occlusion — object hidden then reappears
print("FAILURE MODE 1: Occlusion")
print("  Object goes behind another object.")
print("  Centroid tracker: loses ID after max_disappeared frames, reassigns new ID on return.")
print("  Fix: increase max_disappeared, or use appearance-based re-ID.")

# Failure 2: Fast motion
print("\nFAILURE MODE 2: Fast Motion")
print("  Object moves > max_distance pixels per frame.")
print("  Centroid tracker: can't associate — creates new track, loses old one.")
print("  Fix: increase max_distance, or use Kalman filter to predict future position.")

# Failure 3: ID switching when objects cross paths
print("\nFAILURE MODE 3: ID Switching")
print("  Two objects cross paths at similar speeds.")
print("  Centroid tracker: swaps their IDs at the crossing point.")
print("  Fix: Hungarian algorithm instead of greedy nearest-neighbor matching.")

# Demonstrate: two objects crossing → ID swap
canvas = np.zeros((300,500,3),dtype=np.uint8)
steps  = 25

fig, axes = plt.subplots(1,3,figsize=(18,5))
for ax_idx, step in enumerate([0, 12, 24]):
    frame = canvas.copy()
    # Object A: moves right, Object B: moves left — they cross
    ax_pos = int(50  + step * 16)
    bx_pos = int(450 - step * 16)
    cv2.circle(frame,(ax_pos,150),20,(0,255,0),-1)
    cv2.circle(frame,(bx_pos,150),20,(255,0,0),-1)
    cv2.putText(frame,f'A' if ax_pos<bx_pos else 'A?',(ax_pos-8,158),cv2.FONT_HERSHEY_SIMPLEX,0.6,(255,255,255),2)
    cv2.putText(frame,f'B' if bx_pos>ax_pos else 'B?',(bx_pos-8,158),cv2.FONT_HERSHEY_SIMPLEX,0.6,(255,255,255),2)
    axes[ax_idx].imshow(cv2.cvtColor(frame,cv2.COLOR_BGR2RGB))
    axes[ax_idx].set_title(['Before crossing','At crossing','After — IDs may swap!'][ax_idx])
    axes[ax_idx].axis('off')
plt.suptitle('ID switching is a fundamental problem in centroid tracking',fontsize=12)
plt.show()
