# Lesson 03 — Multi-Object Tracking

```python
import cv2

def multi_track(video_source=0):
    cap = cv2.VideoCapture(video_source)
    ret, frame = cap.read()
    if not ret: return

    trackers = cv2.legacy.MultiTracker_create()
    colors   = []

    print("Draw bounding boxes around objects. Press ENTER after each. Press ESC when done.")
    while True:
        roi = cv2.selectROI('Add object (ENTER=add, ESC=done)', frame, False)
        if roi == (0,0,0,0): break
        trackers.add(cv2.TrackerCSRT_create(), frame, roi)
        colors.append(tuple(int(x) for x in __import__('numpy').random.randint(50,255,3)))
        print(f"  Object {len(colors)} added")

    cv2.destroyAllWindows()

    while True:
        ret, frame = cap.read()
        if not ret: break

        success, boxes = trackers.update(frame)
        for i, (box, color) in enumerate(zip(boxes, colors)):
            if success:
                x,y,w,h = [int(v) for v in box]
                cv2.rectangle(frame,(x,y),(x+w,y+h),color,2)
                cv2.putText(frame,f'Obj{i+1}',(x,y-8),cv2.FONT_HERSHEY_SIMPLEX,0.6,color,2)

        cv2.imshow('Multi-Object Tracker — Q to quit', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'): break

    cap.release(); cv2.destroyAllWindows()

print("Run multi_track(0) to start multi-object tracking on webcam")
