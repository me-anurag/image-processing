# Lesson 02 — OpenCV Built-in Trackers

```python
import cv2
import numpy as np

# Available trackers and their characteristics
trackers_info = {
    'CSRT':  'Accurate, medium speed. Best for most use cases.',
    'KCF':   'Fast, less accurate. Good for fast-moving objects.',
    'MOSSE': 'Fastest, least accurate. Real-time applications.',
}
for name, desc in trackers_info.items():
    print(f"{name:8}: {desc}")

# Usage pattern (run in .py file with webcam or video)
def track_object(video_source=0):
    cap   = cv2.VideoCapture(video_source)
    ret, frame = cap.read()
    if not ret: return

    # User selects ROI with mouse
    roi = cv2.selectROI('Select object to track (ENTER to confirm)', frame, False)
    cv2.destroyWindow('Select object to track (ENTER to confirm)')

    tracker = cv2.TrackerCSRT_create()
    tracker.init(frame, roi)

    while True:
        ret, frame = cap.read()
        if not ret: break

        success, bbox = tracker.update(frame)

        if success:
            x,y,w,h = [int(v) for v in bbox]
            cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),2)
            cx,cy = x+w//2, y+h//2
            cv2.circle(frame,(cx,cy),4,(0,0,255),-1)
            cv2.putText(frame,'TRACKING',(x,y-10),cv2.FONT_HERSHEY_SIMPLEX,0.7,(0,255,0),2)
        else:
            cv2.putText(frame,'LOST',(50,80),cv2.FONT_HERSHEY_SIMPLEX,1.5,(0,0,255),3)

        cv2.imshow('CSRT Tracker — press Q to quit', frame)
        if cv2.waitKey(1) & 0xFF == ord('q'): break

    cap.release(); cv2.destroyAllWindows()

print("\nRun track_object(0) for webcam or track_object('video.mp4') for file")
