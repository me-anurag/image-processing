# Lesson 01 — Binary Thresholding: All Five Types

## Why This Lesson
Thresholding converts a grayscale image into black and white — separating objects from background.
It is the entry point to segmentation. Knowing all five types lets you pick the right one immediately.

## The Five Types
| Type | Pixel > threshold | Pixel ≤ threshold |
|---|---|---|
| THRESH_BINARY | 255 | 0 |
| THRESH_BINARY_INV | 0 | 255 |
| THRESH_TRUNC | threshold | unchanged |
| THRESH_TOZERO | unchanged | 0 |
| THRESH_TOZERO_INV | 0 | unchanged |

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

thresh_val = 127

types = [
    (cv2.THRESH_BINARY,     'BINARY'),
    (cv2.THRESH_BINARY_INV, 'BINARY_INV'),
    (cv2.THRESH_TRUNC,      'TRUNC'),
    (cv2.THRESH_TOZERO,     'TOZERO'),
    (cv2.THRESH_TOZERO_INV, 'TOZERO_INV'),
]

fig, axes = plt.subplots(2, 3, figsize=(18, 11))
axes[0,0].imshow(gray, cmap='gray'); axes[0,0].set_title(f'Grayscale (threshold={thresh_val})'); axes[0,0].axis('off')

for ax, (t, name) in zip(axes.flat[1:], types):
    _, result = cv2.threshold(gray, thresh_val, 255, t)
    ax.imshow(result, cmap='gray'); ax.set_title(name); ax.axis('off')

plt.suptitle('All 5 threshold types on the same image', fontsize=13)
plt.tight_layout(); plt.show()
```

## Key Takeaway
BINARY and BINARY_INV are what you'll use 95% of the time.
The others are useful for specific masking operations — know they exist.
