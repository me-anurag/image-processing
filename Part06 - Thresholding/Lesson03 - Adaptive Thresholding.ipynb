# Lesson 03 — Adaptive Thresholding

## Why This Lesson
Global threshold (including Otsu) applies one value to the whole image.
Adaptive threshold computes a different threshold for each small region.
This handles uneven lighting that ruins global methods.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Global (fails on uneven lighting)
_, global_thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Adaptive Mean: threshold = mean of neighborhood - C
adaptive_mean = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    blockSize=11,   # neighborhood size (must be odd)
    C=2             # constant subtracted from mean
)

# Adaptive Gaussian: threshold = weighted Gaussian mean - C
adaptive_gauss = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    blockSize=11,
    C=2
)

fig, axes = plt.subplots(1, 4, figsize=(22, 6))
for ax, im, t in zip(axes,
    [gray, global_thresh, adaptive_mean, adaptive_gauss],
    ['Grayscale', 'Global Otsu', 'Adaptive Mean', 'Adaptive Gaussian']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
plt.suptitle('Adaptive thresholding handles shadows and uneven lighting', fontsize=12)
plt.show()

# blockSize comparison
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, bs in zip(axes, [7, 11, 21, 51]):
    result = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                   cv2.THRESH_BINARY, bs, 2)
    ax.imshow(result, cmap='gray'); ax.set_title(f'blockSize={bs}'); ax.axis('off')
plt.suptitle('Larger blockSize = adapts to larger lighting variations', fontsize=12)
plt.show()
```

## Key Takeaway
Use adaptive thresholding whenever your image has uneven lighting, shadows, or was photographed
under non-uniform illumination. blockSize controls how local the adaptation is.
