# Lesson 01 — Bounding Boxes: Regular and Rotated

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
se = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
binary = cv2.morphologyEx(binary, cv2.MORPH_OPEN, se)
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
large = [c for c in contours if cv2.contourArea(c) > 500]

vis = img.copy()
for c in large:
    # Regular bounding box (axis-aligned)
    x, y, w, h = cv2.boundingRect(c)
    cv2.rectangle(vis, (x,y), (x+w,y+h), (0,255,0), 2)

    # Rotated bounding box (minimum area)
    rect  = cv2.minAreaRect(c)
    box   = cv2.boxPoints(rect)
    box   = np.int0(box)
    cv2.drawContours(vis, [box], 0, (0,0,255), 2)

    # Enclosing circle
    (cx,cy), radius = cv2.minEnclosingCircle(c)
    cv2.circle(vis, (int(cx),int(cy)), int(radius), (255,0,0), 2)

plt.figure(figsize=(12,6))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title('Green=regular bbox, Red=rotated bbox, Blue=enclosing circle')
plt.axis('off'); plt.show()
```

## Key Takeaway
Regular bbox: fast, axis-aligned. Use for initial detection.
Rotated bbox: minimum area, follows object orientation. Use for measuring actual object size.
Enclosing circle: use for round objects (coins, cells, bubbles).
