# Lesson 03 — Hu Moments: Shape Invariants

## Why
Hu moments describe shape in a way that is invariant to translation, scale, and rotation.
This lets you compare shapes even when they appear at different sizes and angles.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def shape_similarity(c1, c2):
    hu1 = cv2.HuMoments(cv2.moments(c1)).flatten()
    hu2 = cv2.HuMoments(cv2.moments(c2)).flatten()
    # Log transform (Hu moments span many orders of magnitude)
    hu1 = -np.sign(hu1) * np.log10(np.abs(hu1) + 1e-10)
    hu2 = -np.sign(hu2) * np.log10(np.abs(hu2) + 1e-10)
    return np.sum(np.abs(hu1 - hu2))

canvas = np.zeros((300, 800), dtype=np.uint8)
cv2.rectangle(canvas, (50, 50),  (200, 200), 255, -1)   # square
cv2.rectangle(canvas, (250, 50), (430, 230), 255, -1)   # larger square (same shape)
cv2.circle(canvas,    (580, 130), 90, 255, -1)           # circle (different shape)

contours, _ = cv2.findContours(canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
c1, c2, c3  = sorted(contours, key=lambda c: cv2.boundingRect(c)[0])

print(f"Square vs LargerSquare similarity: {shape_similarity(c1,c2):.4f}  (low=similar)")
print(f"Square vs Circle similarity:       {shape_similarity(c1,c3):.4f}  (high=different)")

hu = cv2.HuMoments(cv2.moments(c1)).flatten()
print("\nHu Moments of square:", hu)
```

## Key Takeaway
Low Hu distance = similar shapes. High = different.
Hu moments are scale/rotation/translation invariant — the same shape at any size/angle gives similar values.
