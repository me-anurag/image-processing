# Lesson 02 — Moments and Centroids

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create test shapes
canvas = np.zeros((400, 600), dtype=np.uint8)
cv2.rectangle(canvas, (50, 50), (200, 200), 255, -1)
cv2.circle(canvas, (400, 200), 100, 255, -1)
cv2.ellipse(canvas, (300, 320), (120, 50), 30, 0, 360, 255, -1)

contours, _ = cv2.findContours(canvas, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

vis = cv2.cvtColor(canvas, cv2.COLOR_GRAY2BGR)
for c in contours:
    M  = cv2.moments(c)
    if M['m00'] == 0: continue
    cx = int(M['m10'] / M['m00'])
    cy = int(M['m01'] / M['m00'])

    area      = M['m00']
    perimeter = cv2.arcLength(c, True)

    cv2.drawContours(vis, [c], -1, (0,255,0), 2)
    cv2.circle(vis, (cx,cy), 6, (0,0,255), -1)
    cv2.putText(vis, f'A={area:.0f}', (cx-30,cy+25), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,0), 1)

    print(f"Centroid: ({cx},{cy})  Area: {area:.0f}  Perimeter: {perimeter:.0f}")

plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title('Centroids (red dots) and areas'); plt.axis('off'); plt.show()
```

## Key Takeaway
`M = cv2.moments(contour)` gives all geometric moments.
Centroid: cx = M['m10']/M['m00'], cy = M['m01']/M['m00'].
m00 = area. Always check m00 > 0 before dividing.
