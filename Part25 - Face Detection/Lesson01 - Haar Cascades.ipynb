# Lesson 01 — Haar Cascades

```python
import cv2
import matplotlib.pyplot as plt

# Haar cascades are XML files containing trained classifiers
# Built into OpenCV
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
eye_cascade  = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_eye.xml')

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces
# scaleFactor: how much image is reduced at each pyramid level
# minNeighbors: how many overlapping detections needed before accepting
# minSize: minimum face size to detect
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30,30)
)

vis = img.copy()
for (x,y,w,h) in faces:
    cv2.rectangle(vis,(x,y),(x+w,y+h),(255,0,0),2)
    # Detect eyes only inside face ROI
    roi_gray = gray[y:y+h, x:x+w]
    eyes = eye_cascade.detectMultiScale(roi_gray, 1.1, 3)
    for (ex,ey,ew,eh) in eyes:
        cv2.rectangle(vis,(x+ex,y+ey),(x+ex+ew,y+ey+eh),(0,255,0),2)

print(f"Detected {len(faces)} face(s)")
plt.figure(figsize=(10,6))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title(f'Haar: {len(faces)} faces (blue), eyes (green)'); plt.axis('off'); plt.show()
```

## Key Takeaway
Higher scaleFactor (1.3) = faster but misses faces. Lower (1.05) = more detections, slower.
Higher minNeighbors = fewer false positives but may miss real faces. Start with 5.
