# Lesson 05 — Haar vs DNN: Speed and Accuracy Tradeoffs

```python
import cv2
import numpy as np
import time
import matplotlib.pyplot as plt

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Haar speed
fc = cv2.CascadeClassifier(cv2.data.haarcascades+'haarcascade_frontalface_default.xml')
t0 = time.time()
for _ in range(50): fc.detectMultiScale(gray,1.1,5)
t_haar = (time.time()-t0)/50*1000

print("Comparison:")
print(f"  Haar Cascade:  ~{t_haar:.1f}ms/frame")
print(f"  DNN (SSD):     ~10-30ms/frame (GPU much faster)")
print()
print("Accuracy comparison:")
print("  Haar: good on frontal faces, fails on rotation>30°, many false positives")
print("  DNN:  handles all angles, partial occlusion, much fewer false positives")
print()
print("When to use each:")
print("  Haar: embedded systems, very low power, frontal faces guaranteed")
print("  DNN:  everything else — real-world conditions")

# Demonstrate Haar failure: tilted face
M = cv2.getRotationMatrix2D((img.shape[1]//2,img.shape[0]//2),45,1)
tilted = cv2.warpAffine(img,M,(img.shape[1],img.shape[0]))
tilted_gray = cv2.cvtColor(tilted,cv2.COLOR_BGR2GRAY)
faces_tilted = fc.detectMultiScale(tilted_gray,1.1,5)
print(f"\nHaar on 45° tilted face: {len(faces_tilted)} detections (likely 0)")
```

## Key Takeaway
For any production system use DNN detection. Haar is only for constrained environments
where speed is critical and faces are guaranteed to be frontal.
