# Lesson 02 — Tuning Detection Parameters

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

fc = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

configs = [
    dict(scaleFactor=1.05, minNeighbors=3,  minSize=(20,20)),
    dict(scaleFactor=1.1,  minNeighbors=5,  minSize=(30,30)),
    dict(scaleFactor=1.3,  minNeighbors=8,  minSize=(40,40)),
    dict(scaleFactor=1.5,  minNeighbors=10, minSize=(50,50)),
]

fig, axes = plt.subplots(1,4,figsize=(22,5))
for ax, cfg in zip(axes, configs):
    faces = fc.detectMultiScale(gray, **cfg)
    vis   = img.copy()
    for (x,y,w,h) in faces:
        cv2.rectangle(vis,(x,y),(x+w,y+h),(0,255,0),2)
    ax.imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB))
    ax.set_title(f'sf={cfg["scaleFactor"]}\nmn={cfg["minNeighbors"]} → {len(faces)} faces')
    ax.axis('off')
plt.suptitle('Parameter effect: left=more detections (more false+), right=fewer (may miss)',fontsize=11)
plt.show()
```

## Key Takeaway
If detecting too many false faces: increase minNeighbors, increase scaleFactor.
If missing real faces: decrease both. It is always a tradeoff.
