# Lesson 03 — Eye, Smile, and Body Cascades

```python
import cv2
import matplotlib.pyplot as plt
import os

cascade_dir = cv2.data.haarcascades
print("Available cascades:")
for f in os.listdir(cascade_dir):
    if f.endswith('.xml'): print(f"  {f}")

img  = cv2.imread('sample.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

face_c  = cv2.CascadeClassifier(cascade_dir + 'haarcascade_frontalface_default.xml')
smile_c = cv2.CascadeClassifier(cascade_dir + 'haarcascade_smile.xml')

faces = face_c.detectMultiScale(gray, 1.1, 5, minSize=(60,60))
vis   = img.copy()

for (x,y,w,h) in faces:
    cv2.rectangle(vis,(x,y),(x+w,y+h),(255,0,0),2)
    # Detect smiles only inside face ROI (hierarchical)
    roi  = gray[y:y+h, x:x+w]
    smiles = smile_c.detectMultiScale(roi, 1.7, 20)
    for (sx,sy,sw,sh) in smiles:
        cv2.rectangle(vis,(x+sx,y+sy),(x+sx+sw,y+sy+sh),(0,0,255),2)
        cv2.putText(vis,'SMILE',(x+sx,y+sy-5),cv2.FONT_HERSHEY_SIMPLEX,0.6,(0,0,255),2)

plt.figure(figsize=(10,6))
plt.imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB))
plt.title('Face (blue) + Smile detection inside face ROI (red)'); plt.axis('off'); plt.show()
```

## Key Takeaway
Hierarchical detection: detect face first, then run sub-detectors (eyes, smile) only inside the face ROI.
This drastically reduces false positives and speeds up detection.
