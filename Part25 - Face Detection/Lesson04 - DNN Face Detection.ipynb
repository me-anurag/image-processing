# Lesson 04 — DNN-Based Face Detection

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request, os

# Download model files if not present
proto_url  = 'https://raw.githubusercontent.com/opencv/opencv/master/samples/dnn/face_detector/deploy.prototxt'
weights_url= 'https://github.com/opencv/opencv_3rdparty/raw/dnn_samples_face_detector_20170830/res10_300x300_ssd_iter_140000.caffemodel'

if not os.path.exists('deploy.prototxt'):
    print("Downloading model files...")
    urllib.request.urlretrieve(proto_url,   'deploy.prototxt')
    urllib.request.urlretrieve(weights_url, 'face_model.caffemodel')
    print("Downloaded.")

net  = cv2.dnn.readNetFromCaffe('deploy.prototxt', 'face_model.caffemodel')
img  = cv2.imread('sample.jpg')
h, w = img.shape[:2]

blob = cv2.dnn.blobFromImage(cv2.resize(img,(300,300)), 1.0, (300,300), (104,177,123))
net.setInput(blob)
dets = net.forward()

vis = img.copy()
for i in range(dets.shape[2]):
    conf = dets[0,0,i,2]
    if conf > 0.5:
        box = dets[0,0,i,3:7] * np.array([w,h,w,h])
        x1,y1,x2,y2 = box.astype(int)
        cv2.rectangle(vis,(x1,y1),(x2,y2),(0,255,0),2)
        cv2.putText(vis,f'{conf:.2f}',(x1,y1-8),cv2.FONT_HERSHEY_SIMPLEX,0.6,(0,255,0),2)

plt.figure(figsize=(10,6))
plt.imshow(cv2.cvtColor(vis,cv2.COLOR_BGR2RGB))
plt.title('DNN face detection: more accurate, handles rotation and occlusion'); plt.axis('off'); plt.show()
```

## Key Takeaway
DNN detector is a SSD (Single Shot Detector) trained on faces. It handles tilted, partially occluded,
and non-frontal faces that Haar cascades miss completely.
