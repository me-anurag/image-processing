# Lesson 04 — Cross-Check Matching

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('sample.jpg')
img2 = img1.copy()
M = cv2.getRotationMatrix2D((img1.shape[1]//2,img1.shape[0]//2),30,0.85)
img2 = cv2.warpAffine(img2, M, (img2.shape[1], img2.shape[0]))

orb = cv2.ORB_create(nfeatures=500)
kp1,d1 = orb.detectAndCompute(cv2.cvtColor(img1,cv2.COLOR_BGR2GRAY), None)
kp2,d2 = orb.detectAndCompute(cv2.cvtColor(img2,cv2.COLOR_BGR2GRAY), None)

# Without cross-check
bf_no_cc = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=False)
matches_no = bf_no_cc.match(d1, d2)

# With cross-check: A matches B AND B matches A
bf_cc = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches_cc = bf_cc.match(d1, d2)

print(f"Without cross-check: {len(matches_no)} matches")
print(f"With    cross-check: {len(matches_cc)} matches (mutual best)")

vis = cv2.drawMatches(img1,kp1,img2,kp2,
                      sorted(matches_cc, key=lambda x:x.distance)[:40], None, flags=2)
plt.figure(figsize=(16,5))
plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
plt.title('Cross-check matches — only mutual best matches kept'); plt.axis('off'); plt.show()
```

## Key Takeaway
Cross-check requires A's best match is B AND B's best match is A.
It reduces matches but eliminates many false positives. Use with ORB (BFMatcher + NORM_HAMMING).
Use ratio test with SIFT/FLANN instead.
