# Lesson 05 — Frame Timing and FPS Control

```python
import cv2
import time

cap = cv2.VideoCapture(0)
fps_counter, fps_start, fps_display = 0, time.time(), 0

while True:
    ret, frame = cap.read()
    if not ret: break

    # Processing
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # FPS calculation
    fps_counter += 1
    elapsed = time.time() - fps_start
    if elapsed >= 1.0:
        fps_display = fps_counter / elapsed
        fps_counter, fps_start = 0, time.time()

    # FPS overlay
    cv2.putText(frame, f'FPS: {fps_display:.1f}', (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

    cv2.imshow('Live FPS Counter', frame)

    # waitKey delay controls max FPS:
    # waitKey(1)  = up to 1000 FPS (CPU limited)
    # waitKey(33) = ~30 FPS cap
    # waitKey(0)  = wait forever (pause)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
