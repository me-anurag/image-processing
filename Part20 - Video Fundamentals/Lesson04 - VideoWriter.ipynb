# Lesson 04 — VideoWriter: Saving Processed Video

```python
import cv2

cap = cv2.VideoCapture('sample_video.mp4')
fps    = cap.get(cv2.CAP_PROP_FPS)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# VideoWriter: codec, fps, (width, height)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # use 'XVID' for .avi
out    = cv2.VideoWriter('output.mp4', fourcc, fps, (width, height))

while True:
    ret, frame = cap.read()
    if not ret: break

    # Process each frame
    processed = cv2.cvtColor(cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY), cv2.COLOR_GRAY2BGR)

    # Write to output video
    out.write(processed)

cap.release()
out.release()
print("Saved: output.mp4")
print(f"Note: output size must match VideoWriter size exactly: {width}x{height}")
