# Lesson 03 — Video Properties

```python
import cv2
import matplotlib.pyplot as plt

cap = cv2.VideoCapture('sample_video.mp4')

fps         = cap.get(cv2.CAP_PROP_FPS)
width       = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height      = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration    = frame_count / fps if fps > 0 else 0

print(f"Resolution:   {width} × {height} px")
print(f"FPS:          {fps:.1f}")
print(f"Total frames: {frame_count}")
print(f"Duration:     {duration:.1f} seconds  ({duration/60:.1f} minutes)")

# Jump to a specific frame
cap.set(cv2.CAP_PROP_POS_FRAMES, frame_count // 2)  # middle of video
ret, mid_frame = cap.read()

cap.set(cv2.CAP_PROP_POS_FRAMES, 0)  # jump back to start
ret, first_frame = cap.read()

if ret:
    fig, axes = plt.subplots(1,2,figsize=(14,5))
    axes[0].imshow(cv2.cvtColor(first_frame, cv2.COLOR_BGR2RGB)); axes[0].set_title('Frame 0 (start)'); axes[0].axis('off')
    axes[1].imshow(cv2.cvtColor(mid_frame,   cv2.COLOR_BGR2RGB)); axes[1].set_title('Frame (middle)'); axes[1].axis('off')
    plt.show()

cap.release()
