# Lesson 01 — VideoCapture: Reading from File and Webcam

```python
import cv2
import matplotlib.pyplot as plt

# From FILE
cap_file = cv2.VideoCapture('sample_video.mp4')
ret, frame = cap_file.read()
if ret:
    plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    plt.title('First frame from video file'); plt.axis('off'); plt.show()
cap_file.release()

# From WEBCAM (index 0 = default camera)
# cap_cam = cv2.VideoCapture(0)
# ret, frame = cap_cam.read()
# cap_cam.release()

print("VideoCapture(0)       → default webcam")
print("VideoCapture(1)       → second camera")
print("VideoCapture('file')  → video file (mp4, avi, mov...)")
print("cap.read()            → (success_bool, frame_ndarray)")
print("cap.release()         → ALWAYS call this when done")
