# Lesson 02 — The Frame Loop

```python
import cv2

# The canonical video processing loop
# Run this in a .py file — it won't display in notebooks (no GUI)

cap = cv2.VideoCapture('sample_video.mp4')
frame_count = 0

while True:
    ret, frame = cap.read()

    # ret=False when video ends or camera disconnected
    if not ret:
        print("Video ended or no frame available")
        break

    frame_count += 1

    # --- YOUR PROCESSING GOES HERE ---
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # --- END PROCESSING ---

    cv2.imshow('Video', gray)

    # waitKey(1) = 1ms delay, necessary for imshow to render
    # & 0xFF handles key codes across platforms
    key = cv2.waitKey(1) & 0xFF
    if key == ord('q'):     # Q to quit
        break
    elif key == ord('p'):   # P to pause
        cv2.waitKey(0)

cap.release()
cv2.destroyAllWindows()
print(f"Processed {frame_count} frames")
