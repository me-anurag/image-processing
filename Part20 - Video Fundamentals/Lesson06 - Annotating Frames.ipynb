# Lesson 06 — Annotating Frames in Real Time

```python
import cv2
import numpy as np
import time

# All annotation functions demonstrated on a single frame
img = np.zeros((480, 640, 3), dtype=np.uint8)

# Text
cv2.putText(img, 'Hello World', (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (255,255,255), 2)
cv2.putText(img, 'Status: RUNNING', (50, 90), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

# Rectangle
cv2.rectangle(img, (50, 120), (250, 220), (0,255,0), 2)
cv2.rectangle(img, (50, 120), (250, 220), (0,255,0,50), -1)  # filled (note: no alpha in BGR)

# Circle
cv2.circle(img, (400, 170), 60, (0,0,255), 2)
cv2.circle(img, (400, 170), 5,  (0,0,255), -1)

# Line
cv2.line(img, (0, 300), (640, 300), (255,0,0), 2)

# Arrow
cv2.arrowedLine(img, (300, 350), (450, 350), (255,255,0), 2)

# Overlay transparent info box
overlay = img.copy()
cv2.rectangle(overlay, (10, 400), (630, 470), (50,50,50), -1)
img = cv2.addWeighted(img, 0.7, overlay, 0.3, 0)
cv2.putText(img, f'Frame: demo  |  Time: {time.strftime("%H:%M:%S")}', (20, 445),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (200,200,200), 1)

import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Common annotation functions'); plt.axis('off'); plt.show()
